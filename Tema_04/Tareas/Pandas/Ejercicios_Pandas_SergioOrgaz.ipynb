### Ejercicio 1:

Lee los datos del fichero 'house.csv' en un dataframe. 

- ¿Hay alguna casa con 8 habitaciones?
- ¿Cual es el número mínimo y máximo de habitaciones en el DF?
- Crea una nueva columna que sea la relación de precio por planta. 

#### Solución:

```python
import pandas as pd

# Cargamos el fichero house.csv
df = pd.read_csv('data/house.csv')

# Comprobamos que haya casas con 8 habitaciones
casas8Habitaciones = df[df['bedrooms'] == 8]
print(f'Número de casas con 8 habitaciones: {len(casas8Habitaciones)}')

# Obtenemos el número minimo de habitaciones
numeroMinimoHabitaciones = df['bedrooms'].min()
print(f'Número minimo de habitaciones: {numeroMinimoHabitaciones}')

#Obtenemos el número máximo de habitaciones
numeroMaximoHabitaciones = df['bedrooms'].max()
print(f'Número máximo de habitaciones: {numeroMaximoHabitaciones}')

# Creamos la nueva columna de precio por planta
df['precioPorPlanta'] = df['price']/df['floors']
# Mostramos el dataframe para ver la nueva columna
df.head()
```

### Ejercicio 2

Haciendo uso del mismo DF del ejercicio anterior calcula:
- El precio de la casa que se encuentra en la fila número 256
- El número de habitaciones de las filas 215 a 222
- Recupera una selección aleatoria del 15% del DF
- De ese 15% aleatorio, filtra aquellos registros con 3 o 4 habitaciones y precio menor a 300_000 euros

#### Solución:

```python
import pandas as pd

# Cargamos el fichero house.csv
df = pd.read_csv('data/house.csv')

# Mostramos el precio de la casa que se encuentra en la fila 256
print(f'La casa de la fila 256 tiene un precio de: {df.iloc[256]['price']}')

# Mostramos el número de habitaciones de las casas de la fila 215 hasta la 222
print('Fila:  Número Habitaciones:')
print(df.iloc[215:223]['bedrooms'])

# Obtenemos un 15% aleatorio del dataframe
df15PorCiento = df.sample(frac=0.15)
print(f'DataFrame 15%: {len(df15PorCiento)} filas')

# Filtramos el dataframe del 15% aleatorio para obtener aquellas casa con 3 o 4 habitaciones y un precio
# por debajo de los 300000
casasFiltradas = df15PorCiento[(df15PorCiento['bedrooms'].isin([3, 4])) & (df15PorCiento['price'] < 300000)]
print('Casas Filtradas: ')
casasFiltradas
```

### Ejercicio 3

Genera un índice multiple con las columnas "street", "city", "statezip" y "country" bajo el nombre "localizacion", y el resto de columnas bajo el nombre "caracteristicas". Ordena dichas columnas y genera un nuevo DF que contenga solo las columnas de "localizacion".

#### Solución:

```python
import pandas as pd

# Cargamos el fichero house.csv
df = pd.read_csv('data/house.csv')

# Creamos el indice multiple
localizacion = ['street', 'city', 'statezip', 'country']

# Creamos una tupla que filtre todas las columnas del dataframe, si no pertenerce al indicie de localizacion, se agregan al
# indice de caracteristicas
tuplaIndices = [('localizacion', col) if col in localizacion else ('caracteristicas', col) for col in df.columns]

# Asignamos el multiindice a las columnas
df.columns = pd.MultiIndex.from_tuples(tuplaIndices)

# Ordenamos el dataframe
df = df.sort_index(axis=1)

# Creamos un nuevo dataframe con la columna de localizacion
dfLocalizacion = df['localizacion']

# Mostramos el dataframe
dfLocalizacion.head()
```

### Ejercicio 4

Carga el fichero house.csv y realiza las siguientes operaciones:

- Convierte la columna 'date' en un tipo fecha.
- Genera 3 nuevas columnas fruto de extraer el año, mes y dia.
- ¿Cual es el precio medio por mes?
- Crea una nueva columna llamada **date2** fruto de sumar 20 días a la columna fecha.

#### Solución:

```python
import pandas as pd

# Cargamos el fichero house.csv
df = pd.read_csv('data/house.csv')

# Convertimos el campo data a tipo fecha
df['date'] = pd.to_datetime(df['date'])

# Creamos una nueva columna para año, mes y dia
df['anio'] = df['date'].dt.year
df['mes'] = df['date'].dt.month
df['dia'] = df['date'].dt.day

# Obtenemos el precio medio por mes agrupando por la columna de mes, obteniendo la media del precio
precioMedioMes = df.groupby('mes')['price'].mean().round(2)
print(f'El precio medio por mes es: {precioMedioMes}')

# Creamos la nueva columna date2 a partir de sumar 20 dias a la columna date
df['date2'] = df['date'] + pd.Timedelta(days=20)  

# Mostramos las dos columnas de fechas
df[['date', 'date2']].head()
```

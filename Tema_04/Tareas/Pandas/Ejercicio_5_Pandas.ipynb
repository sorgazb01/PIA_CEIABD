El dataset contiene la siguiente información:

* **AirportID**: Identificador de cada vuelo para un aeropuerto.
* **Name**: Nombre del aeropuerto.
* **City**: Ciudad en la que se encuentra el aeropuerto.
* **Country**: País o territorio en el que se encuentra el aeropuerto.
* **IATA**: Código de asociación internacional de transporte aéreo, código del aeorpuerto.
* **ICAO**: Código de organización civil internacional, código de aeoropuertto.
* **Latitude**: Coordenada del aeropuerto (latitud).
* **Longitude**: Coordenada del aeropuerto (longitud).
* **Altitude**: Altitud del aeropuerto (en pies).
* **Timezone**: Zona horaria.
* **DST**: Código referente al continente (Daylight savings time). Europe (E), A (US/CANADA), S (South America), O (Australia), Z (New Zeeland), N (None), U (Unknown).
* **Tz**: Zona horaria del aeropuerto. Por ejemplo: (America/Los_Angeles).
* **Type**: Tipo de aeropuerto: airport, station, port, unknown.
* **Source**: Fuente de datos.

* 7 **Vuelve a cargar el dataset de modo que se interpreten correctamente los valores nulos (repite el apartado 4, borra las columnas).**

```python
columnas = ['AirportID', 'Name', 'City', 'Country', 'IATA', 'ICAO', 'Latitude', 'Longitude', 'Altitude', 'Timezone', 'DST', 'Tz', 'Type', 'Source']
dfAeropuertos = pd.read_csv(urlDF, header=None, names=columnas, na_values='\\N')

dfAeropuertos.drop(columns=['AirportID', 'Latitude', 'Longitude', 'Altitude'], inplace=True)
dfAeropuertos
```

* 8 **Diseña y desarrolla una función para revisar los value_counts de cada columna.**

```python
def valueCountsColumnas(columnas):
    for columna in columnas:
        print(f"{columna}")
        print(dfAeropuertos[columna].value_counts(dropna=False))
valueCountsColumnas(dfAeropuertos.columns)
```

* 9 **Diseña y desarrolla una función que sobrescriba los valores nulos de las columnas IATA e ICAO por el valor 'DESCONOCIDO'. Diseñala para que te permita aceptar cualquier columna y cualquier valor a sobreescribir.**

```python
def sobrescribirNulos(df, columna, valor):
    df[columna].fillna(valor, inplace=True)

sobrescribirNulos(dfAeropuertos, 'IATA', 'DESCONOCIDO')
sobrescribirNulos(dfAeropuertos, 'ICAO', 'DESCONOCIDO')
dfAeropuertos
```

* 10 **Implementa una función que te ayude a cambiar el tipo de las variables DST y TZ como categórico. Desarrolla tu función para que te permita cambiar cualquier tipo de cualquier columna.**

```python
def cambiarTipo(df, columna, tipo):
    df[columna] = df[columna].astype(tipo)

cambiarTipo(dfAeropuertos, 'DST', 'category')
cambiarTipo(dfAeropuertos, 'Tz', 'category')
dfAeropuertos.dtypes
```

* 11 **Implementa una función que te termina ver en pantalla un resumen estadístico de todas las variables categóricas.**

```python
def resumenCategoricas(df):
    for columna in df.select_dtypes(include='category').columns:
        print(f"{columna}")
        print(df[columna].describe())

resumenCategoricas(dfAeropuertos)
```

* 12 **Diseña y desarrolla una función que te permita agrupar el dataframe por el tipo de aeropuerto, mostrando el conteo de los tipos. Tu función debe estar preparada para agrupar por cualquier campo.**

```python
def agrupar(df, columna):
    return df.groupby(columna).size().reset_index(name='Conteo')

agrupar(dfAeropuertos, 'Type')
```

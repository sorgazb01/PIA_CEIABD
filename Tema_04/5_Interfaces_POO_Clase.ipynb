# Interfaces

En Python, **no existe una forma explícita de definir una interfaz** como en algunos lenguajes de programación orientados a objetos. Sin embargo, podemos **emular el comportamiento de una interfaz usando clases abstractas y métodos abstractos.**

* Las interfaces en Python son un concepto abstracto que **se utiliza para definir un contrato que deben cumplir las clases que implementan una determinada interfaz.** 

* **Una interfaz en Python es una clase que contiene solo definiciones de métodos, pero no tiene implementaciones de estos métodos.** Las clases que implementan una interfaz **deben proporcionar implementaciones** para todos los métodos definidos en la interfaz.

* Por ejemplo, si se quiere definir una interfaz para una clase que representa un objeto geométrico, se podría crear una **interfaz llamada ObjetoGeometrico que contenga métodos como get_area() y get_perimeter().**
    * Las clases que implementen esta interfaz deben proporcionar implementaciones para estos métodos, de manera que se puedan calcular el área y el perímetro de cualquier objeto geométrico.

* En Python, **las interfaces se definen utilizando la palabra clave class,** pero no se implementan como clases normales. 

* La **clase ABC (abreviatura de "Abstract Base Class")** es una clase de la biblioteca estándar de Python que se **utiliza para definir clases abstractas y métodos abstractos.**

    * **La clase ABC proporciona un marco para la creación de clases abstractas y métodos abstractos** en Python. 
    * Al importar la clase ABC, podemos utilizar el **decorador @abstractmethod** para declarar métodos abstractos y la clase ABC se encargará de crear una clase abstracta.

Veamos como se utiliza es sintaxis especial para definir una interfaz:

```python
from abc import ABC, abstractmethod

class ObjetoGeometrico(ABC):
    @abstractmethod
    def get_area(self):
        pass

    @abstractmethod
    def get_perimetro(self):
        pass
```

1. En este ejemplo, se ha importado la clase ABC y el decorador abstractmethod del módulo abc de Python. 

2. La clase ObjetoGeometrico se define como una subclase de ABC, lo que indica que se trata de una interfaz.

3. Se utiliza el **decorador abstractmethod en los métodos get_area() y get_perimeter()** para indicar que se trata de métodos abstractos que deben ser implementados por las clases que implementen esta interfaz.

4. Para implementar una interfaz, se debe crear una clase que herede de la interfaz y proporcionar implementaciones para todos los métodos abstractos definidos en la interfaz. 

5. Para implementar la interfaz GeometricObject se podría crear una clase Circle como se muestra a continuación:

```python
class Circulo(ObjetoGeometrico):
    def __init__(self, radio):
        self.radio = radio

    def get_area(self):
        return 3.14 * self.radio ** 2

    def get_perimetro(self):
        return 2 * 3.14 * self.radio
```

```python
c1 = Circulo(5)
```

```python
print(f"El área del circulo {c1.radio} es: {c1.get_area()} y su perímetro: {c1.get_perimetro()}")
```

En este ejemplo, la clase Circle hereda de la interfaz ObjetoGeometrico y proporciona implementaciones para los métodos get_area() y get_perimeter(). De esta manera, se puede tener una interfaz de la que hereden otras clases e implementar los métodos abstractos definidos en la clase ObjetoGeometrico.

## Ejemplo de Uso. Interfaz para clases que representen objetos serializables

La serialización es el proceso de convertir un objeto en una representación de datos que pueda ser almacenada o transmitida. La deserialización es el proceso inverso de convertir una representación de datos en un objeto.

```python
from abc import ABC, abstractmethod

class Serializable(ABC):
    @abstractmethod
    def to_json(self):
        pass

    @abstractmethod
    def from_json(self, json_data):
        pass
```

```python
class User(Serializable):
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def to_json(self):
        return {
            "name": self.name,
            "email": self.email
        }

    def from_json(self, json_data):
        self.name = json_data["name"]
        self.email = json_data["email"]
```

En este ejemplo, estamos creando una interfaz `Serializable` con métodos abstractos `to_json` y `from_json`. Luego, creamos una clase `User` que implementa la interfaz `Serializable` y proporciona implementaciones para los métodos `to_json` y `from_json`. Esto nos permite **serializar y deserializar** objetos de la clase `User` a formato JSON.

```python
nombre = 'Juan'
email = 'JuanDominguez@gmail.com'
usuario1 = User(email=email, name=nombre)

print(usuario1.to_json())
```

```python
json_data = {'name': 'Pedro', 'email': 'PedroDominguez@gmail.com'}
usuario1.from_json(json_data)
print(usuario1.to_json())
```

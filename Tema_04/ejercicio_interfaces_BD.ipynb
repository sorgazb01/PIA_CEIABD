## Ejercicio.

1. Crea una interfaz para clases que representen adaptadores de base de datos.

   Los métodos comunes para el trabajo con Bases de Datos son:

   1. Conectar.
   2. Ejecutar consulta.
   3. Cerrar conexión.

2. Crea al menos 3 clases que utilicen la interfaz e implementen todos los métodos. No olvides crear los constructores de cada clase.

**Solución:**

### Paso 1 — Definir la interfaz `AdaptadorBD`

Usamos `ABC` y `@abstractmethod` para crear la interfaz con los tres métodos obligatorios:
- `conectar()`: establece la conexión con la base de datos.
- `ejecutar_consulta(consulta)`: ejecuta una consulta SQL o equivalente.
- `cerrar_conexion()`: cierra la conexión abierta.

```python
from abc import ABC, abstractmethod

class AdaptadorBD(ABC):
    """
    Interfaz que deben implementar todos los adaptadores de base de datos.
    Define el contrato con los tres métodos fundamentales de trabajo con BD.
    """

    @abstractmethod
    def conectar(self):
        """Establece la conexión con la base de datos."""
        pass

    @abstractmethod
    def ejecutar_consulta(self, consulta):
        """Ejecuta una consulta sobre la base de datos."""
        pass

    @abstractmethod
    def cerrar_conexion(self):
        """Cierra la conexión con la base de datos."""
        pass
```

### Paso 2 — Implementación 1: `AdaptadorMySQL`

Simula un adaptador para bases de datos **MySQL**. 
El constructor recibe `host`, `usuario`, `contraseña` y `base_de_datos`.

```python
class AdaptadorMySQL(AdaptadorBD):
    """
    Adaptador para bases de datos MySQL.
    Simula la conexión y ejecución de consultas en un servidor MySQL.
    """

    def __init__(self, host, usuario, contrasena, base_de_datos):
        self.host = host
        self.usuario = usuario
        self.contrasena = contrasena
        self.base_de_datos = base_de_datos
        self.conexion = None

    def conectar(self):
        self.conexion = True  # Simulación de conexión real
        print(f"[MySQL] Conectado al servidor '{self.host}', base de datos '{self.base_de_datos}' como '{self.usuario}'.")

    def ejecutar_consulta(self, consulta):
        if not self.conexion:
            print("[MySQL] Error: No hay conexión activa. Llama a conectar() primero.")
            return
        print(f"[MySQL] Ejecutando consulta: {consulta}")

    def cerrar_conexion(self):
        self.conexion = None
        print(f"[MySQL] Conexión cerrada con '{self.host}'.")
```

### Paso 3 — Implementación 2: `AdaptadorPostgreSQL`

Simula un adaptador para **PostgreSQL**.
El constructor recibe `host`, `puerto`, `usuario` y `base_de_datos`.

```python
class AdaptadorPostgreSQL(AdaptadorBD):
    """
    Adaptador para bases de datos PostgreSQL.
    Simula la conexión y ejecución de consultas en un servidor PostgreSQL.
    """

    def __init__(self, host, puerto, usuario, base_de_datos):
        self.host = host
        self.puerto = puerto
        self.usuario = usuario
        self.base_de_datos = base_de_datos
        self.conexion = None

    def conectar(self):
        self.conexion = True
        print(f"[PostgreSQL] Conectado a '{self.host}:{self.puerto}', base de datos '{self.base_de_datos}'.")

    def ejecutar_consulta(self, consulta):
        if not self.conexion:
            print("[PostgreSQL] Error: No hay conexión activa. Llama a conectar() primero.")
            return
        print(f"[PostgreSQL] Ejecutando consulta: {consulta}")

    def cerrar_conexion(self):
        self.conexion = None
        print(f"[PostgreSQL] Conexión cerrada con '{self.host}:{self.puerto}'.")
```

### Paso 4 — Implementación 3: `AdaptadorSQLite`

Simula un adaptador para **SQLite**, una base de datos ligera basada en ficheros.
El constructor solo recibe la `ruta_archivo` donde se almacena la base de datos.

```python
class AdaptadorSQLite(AdaptadorBD):
    """
    Adaptador para bases de datos SQLite.
    Simula la conexión y ejecución de consultas sobre un fichero .db local.
    """

    def __init__(self, ruta_archivo):
        self.ruta_archivo = ruta_archivo
        self.conexion = None

    def conectar(self):
        self.conexion = True
        print(f"[SQLite] Conexión abierta con el archivo '{self.ruta_archivo}'.")

    def ejecutar_consulta(self, consulta):
        if not self.conexion:
            print("[SQLite] Error: No hay conexión activa. Llama a conectar() primero.")
            return
        print(f"[SQLite] Ejecutando consulta: {consulta}")

    def cerrar_conexion(self):
        self.conexion = None
        print(f"[SQLite] Conexión cerrada con '{self.ruta_archivo}'.")
```

### Paso 5 — Prueba de los tres adaptadores

Creamos una instancia de cada adaptador y ejecutamos el flujo completo: conectar → ejecutar consulta → cerrar conexión.

```python
# --- Prueba AdaptadorMySQL ---
mysql = AdaptadorMySQL(
    host="localhost",
    usuario="root",
    contrasena="1234",
    base_de_datos="mi_tienda"
)
mysql.conectar()
mysql.ejecutar_consulta("SELECT * FROM productos WHERE precio < 50;")
mysql.cerrar_conexion()
```

```python
# --- Prueba AdaptadorPostgreSQL ---
postgres = AdaptadorPostgreSQL(
    host="192.168.1.10",
    puerto=5432,
    usuario="admin",
    base_de_datos="ventas"
)
postgres.conectar()
postgres.ejecutar_consulta("SELECT nombre, total FROM pedidos ORDER BY total DESC LIMIT 10;")
postgres.cerrar_conexion()
```

```python
# --- Prueba AdaptadorSQLite ---
sqlite = AdaptadorSQLite(ruta_archivo="datos/inventario.db")
sqlite.conectar()
sqlite.ejecutar_consulta("INSERT INTO articulos (nombre, cantidad) VALUES ('Monitor', 5);")
sqlite.cerrar_conexion()
```

### Nota — Verificación del contrato de la interfaz

Si intentamos instanciar directamente la clase abstracta `AdaptadorBD` sin implementar todos sus métodos, Python lanzará un `TypeError`. Esto garantiza que cualquier adaptador cumpla el contrato definido en la interfaz.

```python
# Intentar instanciar la interfaz directamente genera un error
try:
    adaptador = AdaptadorBD()
except TypeError as e:
    print(f"Error esperado: {e}")
```

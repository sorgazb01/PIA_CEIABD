### 10 Copy - Obtener una Copia de un DataFrame 

Al igual que en listas y arrays los dataframes <font color='orange'>**comparten memoria.** </font>

Es fácil equivocarnos y **asignar a una nueva variable un dataframe completo**, posteriormente, en esta nueva variable **realizar modificaciones,** pensando que sólamente las estamos realizando en la copia del dataframe, **pero no es así, <font color='orange'>estamos realizando modificaciones en ambos dataframes ya que al estar almacenados en memoria, comparten las mismas posiciones,</font> veamos un ejemplo de como NO copiar un dataframe.**


```python
# hacemos un dataframe con los registros que tienen 2 habitaciones
bad_copy = data
bad_copy.head()
```

```python
# Modificamos la primera fila por ceros para probar que la copia no es correcta
bad_copy.iloc[[0]] = 0  # le ponemos 0 a toda la fila

bad_copy.head()
```

**Observamos que los cambios se reflejan en el dataframe original**

```python
data.head(2)
```

**<div class="alert alert-danger">Por lo tanto, hemos modificado ambos dataframes, para únicamente realizar modificaciones sobre la copia. Para hacerlo de forma adecuada, hemos de hacer uso del comando <code>.copy</code></div>**

```python
# cargamos de nuevo los datos originales
data = pd.read_csv('./data/house.csv')

# Copiamos correctamente el dataframe
df_copia = data.copy()

df_copia.iloc[[0]] = 0  # modificamos la copia

df_copia.head(2)
```

```python
# mostramos el df original para comprobar que no ha sido modificado al cambiar la copia.
data.head(2)
```

### 16. Borrar duplicados de un dataframe

**Parámetros:**

Como vemos en el df no hay datos a blanco, por ello **vamos a preparar unos datos NaN a mano con un pequeño código.**

```python
# creamos una tupla compuesta de (numero,precio_vivienda)
tuplas = [(i, v) for i, v in zip(np.arange(len(data)), data['price'])]
tuplas
```

```python
# obtenemos el 5% de los datos para realizar nuestras modificaciones
df1 = data.sample(frac=0.05)
# si el valor de i coincide con el indice de algunos de los valores del 5% seleccionado le ponemos Nan, sino le ponemos el precio
data['price'] = pd.Series(
    [np.nan if i in df1.index.values else v for i, v in tuplas])

# hacemos lo mismo con las habitaciones.
df1 = data.sample(frac=0.05)
tuplas2 = [(i, v) for i, v in zip(np.arange(len(data)), data['bedrooms'])]
data['bedrooms'] = pd.Series(
    [np.nan if i in df1.index.values else v for i, v in tuplas2])
```

## Lección 3 - Conversiones de Datos

### 1. Conversión de Texto a Fecha 

* La librería NumPy de Python define el tipo de dato **datetime64 para trabajar con fechas y diferencias entre fechas.** 


* Un tipo con el que es realmente fácil trabajar con este tipo de información tan complejo, haciendo fáciles operaciones que de otra manera serían complicadas. 

* **Datemime64 es el tipo de dato que se suele usar en los DataFrames de Pandas.**

* A la hora de importar fechas **es habitual que estas se encuentren como cadenas de texto**, por lo que en Pandas existe la función **to_datetime()** mediante la cual se pueden realizar la conversión de estos datos de texto a fecha. 

Veamos el uso básico de esta función.

**pd.to_datetime(arg, errors='raise', dayfirst=False, yearfirst=False, utc=None, format=None, exact=True, unit=None, infer_datetime_format=False, origin='unix', cache=True)**

Siendo los parámetros más importantes:

* **arg:** El elemento que se convierte al tipo de dato fecha, el cual puede ser una serie o un registro único.


* **format:** Una cadena de texto mediante la que se indica el formato en el que se encuentra originalmente la fecha.


* **dayfirst:** Valor lógico mediante el que se indica si el primer valor de la fecha es el día, el valor por defecto de la opción es falso.


* **yearfirst:** Valor lógico mediante el que se indica si el primer valor de la fecha es el año, el valor por defecto de la opción es falso.

#### Conversión básica de cadenas de texto a fecha

Para convertir una cadena de texto en fecha simplemente se tiene que llamar al método to_datetime() pasando como parámetro la cadena de texto. Así una cadena de texto se puede interpretar solamente con

```python
import pandas as pd
pd.to_datetime('10/11/12')
```

* En el ejemplo se ha usado un valor que puede ser ambiguo (“10/11/12”) donde los tres valores se pueden interpretar como el día, el mes o el año de la fecha. 

* La función ha interpretado la fecha como el 11 de octubre de 2012, por lo que ha tomado el primer valor como el mes, el segundo como el día y el tercero como el año. 

* Algo que se puede modificar mediante las propiedades **dayfirst o yearfirst.** En el primero de los casos los valores de la cadena de texto serán interpretados como día, mes y año, mientras que en el segundo como año, mes y día.

```python
print(pd.to_datetime('10/11/12'))                 # 2012-10-11
print(pd.to_datetime('10/11/12', dayfirst=True))  # 2012-11-10
print(pd.to_datetime('10/11/12', yearfirst=True)) # 2010-11-12
```

#### Formato de fecha personalizado con format

Aunque estas opciones pueden ser más que suficiente en muchas ocasiones, es posible tener una mayor presión indicado en formato mediante una cadena de texto con la propiedad format.

```python
pd.to_datetime('10/11/12', format='%d/%y/%m')
```

#### Interpretación de las horas

#### 4.3.1 Codificación Ordinal

Consiste simplemente en **remplazar cada valor de la variable con un número entero distinto.** Es por ello que la codificación ordinal resulta útil cuando **tenemos datos ordinales,** es decir, aquellos en los que **se puede establecer un orden entre sus categóricas.** En consecuencia, esta técnica nos ayudará a representar esta relación de orden de las distintas categorías.

En nuestro conjunto de datos nos centraremos en el atributo ordinal __<code>Nivel_Estudios</code>__, en el cual claramente podemos asignar un orden a los distintos niveles de estudios. 

**Además, conocemos ese orden jerárquico que va desde el nivel de estudios más bajo (Sin_Estudios) al que el algoritmo le asignará un 0, hasta el nivel máximo de educación (Doctor) que tendrá asignado un 5.**

Antes que nada, mostramos todos los valores incluidos en la variable Nivel_Educativo.

```python
#Debemos crear el df_personas en la actividad anterior
df_personas['Nivel_Estudios'].value_counts()
```

Como podemos ver en la tabla superior, hay 6 niveles de educación a los que debemos asignar un número entero. 

Para ello, la librería scikit-learn nos proporciona el objeto **OrdinalEncoder** para aplicar este tipo de codificación y al cual **le podemos indicar el orden de las categorías.**

Veamos cómo podemos utilizarlo en nuestro dataset.

```python
from sklearn.preprocessing import OrdinalEncoder

# Creamos el codificador indicandole el orden de la variables
encoder = OrdinalEncoder(categories=[
                         ["Sin_Estudios", "ESO", "Bachillerato", "Grado_Sup_FP", "Grado_Universitario", "Doctor"]])

# Ajustamos el codificador con la variable education y la transformamos
encoder.fit(df_personas[["Nivel_Estudios"]])
df_personas["Nivel_Estudios_Codificado"] = encoder.transform(
    df_personas[["Nivel_Estudios"]])
```

#### 3.3.2 Codificación One-Hot o Columnas Dummy - Convertir Variables Categoricas Nominales a Numericas

### 1.2 Datos Distribuidos

En muchas ocasiones nos encontraremos los ficheros fraccionados y para desarrollar nuestros modelos, tendremos que unificarlos en un único archivo. 


Nos han dejado 332 ficheros en el directorio./data/distributed-data con algunas caracteristicas adicionales de los vinos.

Nos han encargado que carguemos los datos de todos esos ficheros y los unifiquemos en un nuevo dataframe. Estan son algunas pautas que pueden ayudarnos a resolver la cuestión:

* Importar el primer fichero y sobre este ir añadiendo datos.
* Hacer un bucle para ir recorriendo todos y cada uno de los ficheros. 
    * Importante tener una consistencia en el nombre de los ficheros 
    * Importamos los ficheros uno a uno
    * Cada uno de ellos debe apendizarse (añadirse al final) del primer fichero que ya habíamos cargado
* Repetir el bucle hasta que no queden ficheros.

Desarrolla una función para llevar a cabo la carga de estos ficheros.

```python


def carga_distribuida(filepath: str):
    '''
    Carga de ficheros distribuidos en un único dataframe
    
    Parameters:
    -----------
    filepath: ruta donde se encuentran los ficheros.
    
    Returns:
    --------
    DataFrame con los datos de todos los ficheros 
    
    '''
    data = pd.read_csv(filepath + "/001.csv")    
    final_length = len(data)

    for i in range(2,333):
        if i < 10:
            filename = "00" + str(i)
        if 10 <= i < 100:
            filename = "0" + str(i)
        if i >= 100:
            filename = str(i)
        file = filepath + filename + ".csv"

        temp_data = pd.read_csv(file)
        final_length += len(temp_data)

        data = pd.concat([data, temp_data], axis = 0)
        
    return data
```

```python
df.pivot_table(index = "foo", columns = "bar", values = "baz")
```

Podemos aplicar otra función de agregación utilizando el parámetro **aggfunc:**

```python
#cuenta el nº de veces que aparece cada elemento, lógicamente, solo se repite 1 vez en este df prequeño.
df.pivot_table(index = "foo", columns = "bar", values = "baz", aggfunc = "count")
```

En este ejemplo hemos contado el número de registros representados en cada intersección.

Es posible **aplicar más de una función de agregación a los datos.** En el siguiente ejemplo aplicamos tanto la función de cálculo del **valor medio como el recuento:**

```python
df.pivot_table(index = "foo", columns = "bar", values = "baz", aggfunc = [np.mean, "count"])
```

## Lección 6 - Agrupaciones

"Group by" es un proceso que involucra uno o más de los siguientes pasos: 
- Dividir los datos en grupos según algunos criterios. 
- Aplicar una función a cada grupo de forma independiente. 
- Combinar los resultados en una estructura de datos.

Este proceso de agrupación y agregación es muy similar al que efectuamos en SQL.

```python
import numpy as np
import pandas as pd
```

```python
#lista que usaremos para el genero y sexo y estatus social
genero = ["Masculino", "Femenino"]
estatus = ["Pobre", "Clase Media", "Rico"]
```

```python
n = 500

genero_data = [ np.random.choice(genero) for _ in range(0,n) ]
estatus_data = [ np.random.choice(estatus) for _ in range(0,n) ]


```

```python
genero_data[1:10]
```

```python
estatus_data[1:10]
```

```python
#generar altura, peso, edad e ingresos de forma aleatoria.
altura = np.ceil(160 + 30 * np.random.randn(n))
peso = np.ceil(65 + 25 * np.random.randn(n))
edad = np.ceil(30 + 12 * np.random.randn(n))
ingresos = np.ceil(18000 + 3500 * np.random.rand(n))
```

```python
data = pd.DataFrame(
    {
        "Genero" : genero_data,
        "Estatus_Economico" : estatus_data,
        "Altura" : altura,
        "Peso" : peso,
        "Edad" : edad,
        "Ingresos" : ingresos
    }
)
```

```python
data.head()
```

---

Vamos a ver ahora como hacer agrupaciones con los datos generados

### 1.  Agrupación de datos - Group_By( )

Para ver este bloque de contenidos vamos a crear nuestro propio DataFrame con datos socioeconómicos generados al azar.

```python
grupo_genero.indices['Femenino']
```

```python
doble_grupo = data.groupby(["Genero", "Estatus_Economico"])
```

```python
len(doble_grupo)
```

```python
for nombres, grupos in doble_grupo:
    print(nombres)
    print(grupos)
```

---
### 2. Operaciones Sobre Datos Agrupados

Para aplicar operaciones sobre grupos de datos agrupados tenemos 2 métodos, **agg y aggregate.** Basicamente los dos hacen lo mismo, podemos usarlos indistintamente.

```python
doble_grupo.sum()
```

```python
doble_grupo.mean()
```

```python
doble_grupo.size()
```

```python
doble_grupo.describe()
```

```python
#el agrupado doble, puedo quedarme por las columnas que quiera
grupo_ingresos = doble_grupo.Ingresos
grupo_ingresos.describe()
```

```python
#podemos aplicar distintas operaciones a las columnas
doble_grupo.agg(
    {
        "Ingresos": np.sum,
        "Edad" : np.mean,
        "Altura" : np.std
    }
)
```

```python
#incluso aplicar varias y lambdas
doble_grupo.agg(
    {
        "Edad" : [max, min, np.mean],
        "Altura" : lambda h:(np.mean(h))/np.std(h)
    }
)
```

```python
#aplico 3 funciones, sumatorio, media y desviación estandard.
doble_grupo.agg([np.sum, np.mean, np.std])
#como podemos observar para cada variable numérica se aplican las 3 funciones
```

```python
#también podemos aplicar una función lambda a todas las variables.
doble_grupo.agg([lambda x: np.mean(x) / np.std(x)])
```

### 3. Filtros sobre Agrupaciones

Podemos aplicar también filtros sobre nuestras agrupaciones.

```python
doble_grupo["Edad"].filter(lambda x: x.sum()>2400)
```

```python
doble_grupo.filter(lambda x: x['Edad'].sum()>2400)
```

**También podemos usar el método apply para personalizar la funcion que deseemos.**

```python
data.groupby(['Genero', 'Estatus_Economico'])['Ingresos'].apply(suma_entre_media)
```

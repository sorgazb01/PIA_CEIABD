# Normalización y Estandarización.

* La normalización y estandarizacion **son técnicas útiles de preprocesamiento de datos** porque **pueden ayudar a mejorar el rendimiento** de muchos modelos de aprendizaje automático. 

* Por ejemplo, algunos modelos de aprendizaje automático **pueden ser sensibles a la escala de los datos y pueden rendir mejor cuando los datos están reescalados o normalizados.** 

* También pueden ser útiles para **hacer que los datos sean más fáciles de visualizar o comparar.**

---
## Pautas que pueden ayudarte a decidir si debes utilizar la estandarización o la normalización

**Utiliza la normalización si:**

1. Necesitas que los valores de las características o variables estén en el rango de valores (0, 1).

2. Si los algoritmos de aprendizaje automático que estás utilizando son sensibles a la escala de las características o variables.

**Utiliza la estandarización si:**

1. Necesitas que los valores de las características o variables tengan **media 0 y desviación estándar 1.**

2. Si las características o variables tienen **diferentes unidades de medida o tienen una escala muy diferente.**

3. Si las **características o variables tienen distribuciones muy diferentes.**

4. Si los algoritmos de aprendizaje automático que estás utilizando son sensibles a la escala de las características o variables.

A continuación vamos a estudiar como aplicar Normalización y estandarización a nuestros datos.

# Normalización

* La normalización es una **técnica de preprocesamiento de datos que consiste en ajustar los valores de los datos a un rango específico.** 

* Por ejemplo, es común reescalar los datos para que estén en el rango de **0 a 1.** Esto se puede hacer utilizando la siguiente fórmula:

**`x_rescaled = (x - x_min) / (x_max - x_min)`**


Donde **`x` es el valor original, `x_min` es el valor mínimo del conjunto de datos y `x_max` es el valor máximo del conjunto de datos.** Esta fórmula ajusta los valores de x para que estén en el rango de 0 a 1.

```python
import numpy as np
# Creamos una matriz de datos X con 3 caracteristicas numéricas(columnas)
x = np.array([[2, 3, 5],
              [6, 2, 4],
              [1, 4, 3],
              [8, 5, 1],
              [3, 6, 2],
              [5, 7, 4],
              [9, 3, 6],
              [7, 2, 5],
              [4, 1, 3],
              [2, 8, 4]], dtype=np.float32)
```

```python


#Queremos normalizar cada caracteristica de la matriz (cada columna).
#Para ello tenemos que recorrer cada columna, sacar el minimo, el máximo y luego aplicar el metodo interp
#El reescalado sigue la siguiente formula: x_rescaled = (x - x_min) / (x_max - x_min).

# Recorremos cada columna de la matriz X
for i in range(x.shape[1]):
    # Calculamos el valor mínimo y máximo de la columna
    x_min = np.min(x[:, i])
    x_max = np.max(x[:, i])
    
    # Normalizamos la columna
    x[:, i] = np.interp(x[:, i], (x_min, x_max), (0, 1))

# Si quisiera redondear, podria:
#para cada fila le aplico un map con una función lambda que redonde a 4 decimales.
#a modo ejercicio lo podemos hacer pero en los reescalados y normalizaciones no hay que hacerlo.
#x = [ list( map(lambda x: round(x,4),fila) ) for fila in x ]
```

El código anterior normalizará cada columna de la matriz x para que esté en el rango de 0 a 1. El resultado final sería una matriz x reescalada, que se vería así:

# Estandarización

* La Estandarización es otra técnica de preprocesamiento de datos que **consiste en transformar los datos de tal manera que tengan una distribución normal (o gaussiana)** con una media de 0 y una desviación estándar de 1. 

* Esto se puede hacer utilizando la siguiente fórmula:

**x_estandarizado = (x - x_mean) / x_std**

* Donde x es el valor original, `x_mean` es la media del conjunto de datos y `x_std` es la desviación estándar del conjunto de datos. Esta fórmula ajusta los valores de x para que tengan una media de 0 y una desviación estándar de 1.

Para estandarizar los datos, podemos utilizar la fórmula de estandarización mencionada anteriormente y aplicarla a cada columna de la matriz X de la misma manera que hicimos con la normalización. 

**Vamos a ver un ejemplo de código que realiza la estandarización de los datos:**

```python
import numpy as np

# Creamos una matriz de datos X
X = np.array([[2, 3, 5],
              [6, 2, 4],
              [1, 4, 3],
              [8, 5, 1],
              [3, 6, 2],
              [5, 7, 4],
              [9, 3, 6],
              [7, 2, 5],
              [4, 1, 3],
              [2, 8, 4]], dtype=np.float32)

# Recorremos cada columna de la matriz X
for i in range(X.shape[1]):
    # Calculamos la media y la desviación estándar de la columna
    x_mean = np.mean(X[:, i])
    x_std = np.std(X[:, i])
    
    # estandarizamos la columna
    X[:, i] = (X[:, i] - x_mean) / x_std

print(X)
```

# ¿ Es conveniente aplicar redondeos a las variables o caracteristicas tras realizar una estandarización o normalización?

En general, **no es conveniente redondear los valores de las características o variables tras realizar una estadarización o normalización, ya que el redondeo puede afectar la precisión y la calidad de los resultados.**

La estandarización y la normalización **son técnicas utilizadas para transformar los valores de las características o variables** de un conjunto de datos de manera que **cumplan ciertas condiciones,** como tener un rango de valores específico o tener una media y una desviación estándar determinadas. **Al redondear los valores de las características o variables,estamos alterando los valores originales y, por tanto, también estamos alterando la transformación realizada.**

En cambio, si queremos trabajar con valores redondeados, podemos hacerlo antes de aplicar la estandarización o la normalización, utilizando la función `numpy.round()` o cualquier otra función de redondeo. De esta manera, estamos trabajando con valores redondeados desde el principio y no afectamos la precisión de la transformación realizada.

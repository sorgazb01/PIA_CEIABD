* **Cambiar tamaño, resolución, crear varios gráficos**

El método **figure** en Matplotlib se utiliza para **crear una nueva figura**, es decir, una ventana o página en la interfaz gráfica donde se pueden dibujar uno o varios gráficos. Al crear una figura, puedes especificar varias opciones, como el tamaño de la figura (a través del argumento figsize) y la resolución DPI (dpi).

**subplot**

El método **subplot** en Matplotlib es una **herramienta esencial para crear múltiples gráficos dentro de una misma ventana o figura.** Permite organizar los gráficos en un formato de cuadrícula, **especificando el número de filas y columnas,** así como la posición del gráfico actual en esa cuadrícula.

La sintaxis básica de subplot es:

**plt.subplot(nrows, ncols, index)**


* nrows: Número de filas en la cuadrícula.

* ncols: Número de columnas en la cuadrícula.

* index: Índice del gráfico actual, empezando por 1 en la esquina superior izquierda y aumentando hacia la derecha.


Por ejemplo, si queremos crear una figura con 2 gráficos, uno al lado del otro, usamos subplot de la siguiente manera:

**Ejemplo 1: Dos Gráficos en Una Figura**

```python
import matplotlib.pyplot as plt

# Crear una nueva figura con un tamaño específico
plt.figure(figsize=(10, 4), dpi=300)

# Primer gráfico: línea
plt.subplot(1, 2, 1)# 1 fila, 2 columnas, índice 1 (primera posición)
plt.plot([1, 2, 3, 4], [1, 4, 9, 16], 'ro-')# 'ro-' significa círculos rojos con línea
plt.title('Cuadrados')

# Segundo gráfico: dispersión
plt.subplot(1, 2, 2)# 1 fila, 2 columnas, índice 2 (segunda posición)
plt.scatter([1, 2, 3, 4], [1, 4, 9, 16], color='blue')
plt.title('Dispersión')

# Ajustar el layout y mostrar la figura
plt.tight_layout()
plt.show()
```

subplot también se puede utilizar en un formato más compacto, combinando los tres números nrows, ncols, e index en un solo argumento entero. 

Por ejemplo, **plt.subplot(121) es equivalente a plt.subplot(1, 2, 1).**

**Ejemplo 2: Cuatro Gráficos en Una Figura**

Este ejemplo crea una figura con cuatro gráficos dispuestos en una cuadrícula 2x2:

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos para los gráficos
x = np.arange(0., 5., 0.2)
y = x ** 2

# Crear una nueva figura
plt.figure(figsize=(10, 8), dpi=300)

# Primer gráfico: línea sólida
plt.subplot(2, 2, 1)
plt.plot(x, y, 'r-')
plt.title('Línea sólida')

# Segundo gráfico: línea punteada
plt.subplot(2, 2, 2)
plt.plot(x, y, 'g--')
plt.title('Línea punteada')

# Tercer gráfico: línea punteada con puntos
plt.subplot(2, 2, 3)
plt.plot(x, y, 'b-.')
plt.title('Punteada con puntos')

# Cuarto gráfico: cuadrados
plt.subplot(2, 2, 4)
plt.plot(x, y, 'ks')
plt.title('Cuadrados')

# Ajustar el layout y mostrar la figura
plt.tight_layout()
plt.show()
```

Usamos **plt.figure()** para iniciar una nueva figura, donde se pueden especificar opciones como el tamaño y los dpis. Luego, **plt.subplot()** se emplea para añadir gráficos individuales a la figura en las posiciones deseadas de la cuadrícula.

**subplots**

Es un método que **crea una figura y un conjunto de subgráficos (ejes) en una sola llamada.**

Devuelve una figura y un array de ejes (subgráficos), lo que facilita la manipulación y el acceso a cada subgráfico individualmente a través de este array.

```python
# Ejemplo de uso de subplots
import matplotlib.pyplot as plt

# Devuelve una figura, y un vector(axs) de figuras, en este caso: 1 fila, 2 columnas
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].plot([1, 2, 3], [1, 4, 9])#accedo por el índice a cada posición y pinto el gráfico
axs[1].plot([1, 2, 3], [2, 3, 4])#accedo por el índice a cada posición y pinto el gráfico

plt.show()
```

Vamos a hacer el ejemplo de los 4 gráficos usando subplots

```python
import matplotlib.pyplot as plt
import numpy as np

# Datos para los gráficos
x = np.arange(0., 5., 0.2)
y = x ** 2

# Títulos para cada subgráfico
t = ['Línea sólida', 'Línea punteada', 'Punteada con puntos', 'Cuadrados']

# Crear una nueva figura y vector de figuras
fig, axs = plt.subplots(2, 2, figsize=(10, 8), dpi=300)# dos filas 2 columnas, 2x2

# Primer gráfico: línea sólida
axs[0,0].plot(x, y, 'r-', label=t[0])
axs[0,0].set_title(t[0])
axs[0,0].legend()#ponemos leyenda

# Segundo gráfico: línea punteada
axs[0,1].plot(x, y, 'g--', label=t[1])
axs[0,1].set_title(t[1])
axs[0,1].legend()#ponemos leyenda

# Tercer gráfico: línea punteada con puntos
axs[1,0].plot(x, y, 'b-.')
axs[1,0].set_title(t[2])

# Cuarto gráfico: cuadrados
axs[1,1].plot(x, y, 'ks')
axs[1,1].set_title(t[3])

# Ajustar el layout y mostrar la figura
plt.tight_layout()
plt.show()
```

**Importante**

Mientras que **subplot** se utiliza para **añadir subgráficos individualmente a una figura**, **subplots crea todos los subgráficos de una vez**, devolviendo un array de ejes que se pueden manipular fácilmente. Esto hace que subplots sea generalmente más conveniente para trabajar con múltiples gráficos de manera eficiente.

**1. Partículas Pequeñas Suspendidas en el aire (PM2.5)**

Vamos a simular un conjunto de datos de PM2.5 para una semana. Luego, crearemos un gráfico de líneas para visualizar estos datos, personalizando el gráfico con títulos, etiquetas, añadiendo una línea que indique el límite recomendado por la Organización Mundial de la Salud (OMS) para la concentración de PM2.5, que es de 25 µg/m³ en promedio diario.

```python
#Preparamos los datos
import matplotlib.pyplot as plt
import numpy as np

# Simulando datos diarios de PM2.5 para una semana
dias = ["Lunes", "Martes", "Miércoles", "Jueves", "Viernes", "Sábado", "Domingo"]
pm25 = np.random.randint(10, 50, size=len(dias))  # Valores aleatorios entre 10 y 50
```

```python
# Crear el gráfico
plt.figure(figsize=(10, 6))  # Tamaño del gráfico

plt.plot(dias, pm25, marker='o', linestyle='-', color='blue', label='PM2.5')
plt.axhline(25, color='red', linestyle='--', label='Límite OMS')  # Línea del límite OMS

# Personalización
plt.title("Niveles de PM2.5 durante una Semana")
plt.xlabel("Día")
plt.ylabel("Concentración de PM2.5 (µg/m³)")
plt.legend()
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

**2. Comparación de Temperatura y Precipitación Mensual**

Imaginemos que queremos comparar la temperatura promedio y la precipitación total mensual a lo largo de un año en una ciudad específica. Usaremos datos simulados para la temperatura y la precipitación.

```python
#Preparamos los datos
import matplotlib.pyplot as plt
import numpy as np

# Meses del año
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic']

# Datos simulados
temperaturas = np.random.uniform(10, 30, size=len(meses))  # Temperaturas entre 10 y 30 grados
precipitaciones = np.random.uniform(20, 100, size=len(meses))  # Precipitaciones entre 20 y 100 mm
```

```python
#Crear el gráfico combinado
fig, ax1 = plt.subplots(figsize=(12, 8),dpi= 300)

# Eje para las temperaturas
color = 'tab:red'
ax1.set_xlabel('Mes')
ax1.set_ylabel('Temperatura (°C)', color=color)
ax1.plot(meses, temperaturas, color=color, marker='o', label='Temperatura')
ax1.tick_params(axis='y', labelcolor=color)

# Crear un segundo eje para las precipitaciones
ax2 = ax1.twinx()  # Instancia un segundo eje que comparte el mismo eje X
color = 'tab:blue'
ax2.set_ylabel('Precipitación (mm)', color=color)
ax2.bar(meses, precipitaciones, color=color, alpha=0.6, label='Precipitación')
ax2.tick_params(axis='y', labelcolor=color)

# Título y personalización
plt.title('Temperatura y Precipitación Mensual')
fig.tight_layout()  # Ajusta el layout para que no haya contenido recortado

# Mostrar leyendas
ax1.legend(loc='upper left')
ax2.legend(loc='upper right')


plt.show()
```

* **Distribución de cartera de inversiones:** 
Un gráfico circular para mostrar la distribución de una cartera de inversiones entre diferentes clases de activos (acciones, bonos, efectivo, etc.).


**Definir la Distribución de la Cartera**

Podemos leer de un fichero o crear los datos. 

Supongamos que la distribución de nuestra cartera de inversiones es la siguiente:

Acciones: 40%
Bonos: 20%
Bienes Raíces: 15%
Commodities: 10%
Efectivo: 15%

Creación del gráfico por sectores

```python
import matplotlib.pyplot as plt

# Etiquetas para las diferentes partes de nuestra cartera
labels = ['Acciones', 'Bonos', 'Bienes Raíces', 'Commodities', 'Efectivo']

# Tamaños para cada segmento
sizes = [40, 20, 15, 10, 15]

# Colores para cada segmento
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0']

# Separar el segmento de 'Acciones' para destacarlo
explode = (0.1, 0, 0, 0, 0)# solo "explode" el primer segmento

# Crear el gráfico circular
plt.figure(figsize=(10, 10))
#autopct='%1.1f%%' se utiliza para el formato de los decinales
#startangle especifica el ángulo de inicio para el primer sector
plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', startangle=140)

plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.title('Distribución de Cartera de Inversiones')
plt.show()
```

2. Análisis de Datos de Salud
Tasa de incidencia de enfermedades: Gráficos de barras para comparar la tasa de incidencia de una enfermedad específica a lo largo de diferentes regiones o países.

## 2.1 Configuración de Parámetros

* **1. Títulos de Gráficos, Leyendas y Etiquetas de Ejes**

El título del gráfico y las etiquetas de los ejes y leyenda son fundamentales para proporcionar contexto sobre lo que está representando tu visualización. Vamos a ver cómo personalizarlos:

**Ejemplo con Múltiples Series de Datos**

Imaginemos que queremos comparar las ventas de dos productos a lo largo de varios meses. Esto nos da la oportunidad de trabajar con múltiples series de datos y ver cómo Plotly maneja las leyendas automáticamente.

1. **Preparación de Datos:** Utilizamos un DataFrame de pandas para organizar nuestros datos, lo cual es una práctica común al trabajar con Plotly, especialmente con datos complejos. Este DataFrame incluye una columna para los meses, una para las ventas, y una para identificar el producto (A o B).

```python
import plotly.express as px

# Datos de ejemplo
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo"]
ventas_producto_A = [150, 210, 250, 310, 370]
ventas_producto_B = [140, 200, 240, 280, 320]

# Creamos un DataFrame para manejar nuestros datos más fácilmente
import pandas as pd
df = pd.DataFrame({
    "Mes": meses * 2,# Repetimos los meses para cada producto
    "Ventas": ventas_producto_A + ventas_producto_B,# Concatenamos las ventas
    "Producto": ["A"] * 5 + ["B"] * 5# Etiquetas para cada serie de datos
})
```

```python
df 
```

2. **Gráfico de Líneas Múltiples:** Al pasar **color="Producto" a px.line, Plotly automáticamente distingue las series de datos por producto**, asignando diferentes colores a cada uno y añadiendo una leyenda correspondiente.



```python
# Crear un gráfico de líneas con múltiples series
fig = px.line(df, x="Mes", y="Ventas", color="Producto", title="Ventas por Mes de Productos A y B")
```

3. **Personalización de Títulos y Leyendas:** Con **update_layout, personalizamos los títulos de los ejes y de la leyenda.** **xaxis_title** y **yaxis_title** nos permiten definir títulos claros para los ejes, mientras que **legend_title** modifica el título de la leyenda.

```python



# Personalizar los ejes y la leyenda
fig.update_layout(
    xaxis_title="Mes del Año",
    yaxis_title="Unidades Vendidas",
    legend_title="Producto"
)

# Mostrar el gráfico
fig.show()


```

**Curiosidad**

Si pinchas en la leyenda sobre alguna de las categorías, las puedes activar/desactivar interactivamente.





* **2. Tamaños, fuentes y colores de los títulos, subtítulos, leyendas y etiquetas de los ejes**


Podemos utilizar el método **update_layout junto con update_xaxes y update_yaxes** para los ejes específicamente. Estas funciones permiten una **personalización detallada** de varios aspectos del gráfico.

Vamos a ver cómo ajustar estos elementos en un gráfico, continuando con el ejemplo de las ventas de productos. Personalizaremos el tamaño, la fuente y el color del título del gráfico, así como de las leyendas y las etiquetas de los ejes.

```python
import plotly.express as px
import pandas as pd

# Datos de ejemplo
meses = ["Enero", "Febrero", "Marzo", "Abril", "Mayo"]
ventas_producto_A = [150, 210, 250, 310, 370]
ventas_producto_B = [140, 200, 240, 280, 320]
ventas_producto_C = [500, 300, 350, 500, 620]

# Preparar DataFrame
df = pd.DataFrame({
    "Mes": meses * 3,
    "Ventas": ventas_producto_A + ventas_producto_B + ventas_producto_C,
    "Producto": ["A"] * 5 + ["B"] * 5 + ["C"] * 5
})
```

```python
df
```

```python
# Crear gráfico
fig = px.line(df, x="Mes", y="Ventas", color="Producto", title="Ventas por Mes de Productos")
```

```python

# Personalizar gráfico
fig.update_layout(
    title={
        'text': "Ventas por Mes de Productos",
        'y':0.9,
        'x':0.5,
        'xanchor': 'center',
        'yanchor': 'top',
        'font': {
            'family': "Arial",
            'size': 24,
            'color': "DarkBlue"
        },
    },
    legend_title_text='Producto',
    legend=dict(
        title_font_family="Times New Roman",
        title_font_color="purple",
        title_font_size=16,
        font=dict(
            family="Courier",
            size=12,
            color="black"
        ),
    ),
    xaxis_title="Mes del Año",
    yaxis_title="Unidades Vendidas",
)

# Personalizar ejes
fig.update_xaxes(title_font=dict(family="Helvetica", size=18, color="RebeccaPurple"))
fig.update_yaxes(title_font=dict(family="Helvetica", size=18, color="RebeccaPurple"))

# Mostrar gráfico
fig.show()
```

**2. Histogramas**

Útiles para visualizar la distribución de una variable numérica.

```python
import plotly.express as px
import numpy as np

# Generando datos de muestra: edades
edades = np.random.randint(18, 65, size=200)

# Creando un histograma con un número específico de bins
fig = px.histogram(edades, nbins=20, title='Distribución de Edades')
fig.show()

```

* El parámetro **nbins** en Plotly se utiliza para **especificar el número de intervalos (bins) en los que se desea dividir los datos** al crear histogramas o gráficos de densidad.

* Al crear un histograma, **nbins determina cuántas barras (o "bins") se utilizarán** para agrupar el rango completo de los datos. 

* En este ejemplo, **nbins=20 significa que las edades se dividirán en 20 intervalos iguales para el histograma.** Cambiar el valor de nbins afectará directamente a cómo se visualiza la distribución de las edades, permitiendo una exploración más detallada o más general según sea necesario.



**Consideraciones al Elegir nbins**

* **Datos Discretos vs. Continuos:** 

* Para datos **continuos, ajustar nbins es más crítico,** ya que afecta directamente cómo se agrupan los valores continuos.

* Para **datos discretos** con un número limitado de valores únicos, el ajuste de nbins puede ser **menos relevante.**

**5. Mapas de Calor (Heatmaps)**

* Los mapas de calor (heatmaps) se utilizan frecuentemente para **visualizar relaciones entre dos variables en un conjunto de datos, especialmente para destacar correlaciones, patrones o áreas de intensidad dentro de los datos.** 


* Los mapas de calor pueden ser particularmente útiles para **explorar matrices de correlación, donde los colores representan la fuerza y la dirección de las correlaciones entre las variables.**




Vamos a crear un ejemplo más avanzado y relevante que ilustre cómo se puede usar un mapa de calor para visualizar la correlación entre diferentes variables numéricas en un conjunto de datos.

Usaremos el conjunto de datos de Iris,  que incluye medidas de las características de las flores Iris de tres especies diferentes. Calcularemos la matriz de correlación entre estas características y luego la visualizaremos usando un mapa de calor.

```python
import plotly.express as px
import pandas as pd

# Cargar el conjunto de datos de Iris
df = px.data.iris()

# Seleccionar solo las columnas numéricas para calcular la correlación
df_numeric = df.select_dtypes(include=['float64', 'int64'])

# Calcular la matriz de correlación solo con las columnas numéricas
correlacion = df_numeric.corr()

# Definir una paleta de colores personalizada
color_scale = ["blue", "green", "yellow"]  # De azul a verde a amarillo

# Visualizar la matriz de correlación con un mapa de calor
fig = px.imshow(correlacion, text_auto=True, aspect="auto",
                labels=dict(x="Característica", y="Característica", color="Correlación"),
                x=correlacion.columns,
                y=correlacion.columns,
                color_continuous_scale='ylgnbu',  # Usar la escala de colores personalizada
                title="Mapa de Calor de la Correlación entre Características de Iris")

fig.show()

```

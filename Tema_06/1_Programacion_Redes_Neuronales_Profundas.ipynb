## **3. Tipos de Redes Neuronales**

A dia de hoy, tenemos 5 tipos de Redes Neuronales Artificiales (ANN - Artificial Neural Networks)

1. **Perceptron Multicapa o Multilayer Perceptron (MLP)**
2. **Redes Convolucionales (ConvNets)**
3. **Redes Recurrentes (RNNs)**
4. **Redes Autoencoder**
5. **Redes Generativas Adversarias (GANs)**

Vamos a ver con un poco más de detalle en qué consiste cada una de ellas.

### **3.1 Perceptron Multicapa o Multilayer Perceptron (MLP):** 

Son redes neuronales que **tienen una estructura en forma de un árbol, donde las entradas fluyen hacia adelante y hacia una capa de salida.** Estas redes están recomendadas para problemas de **clasificación y regresión**, como por ejemplo, la clasificación de imágenes o la predicción de precios de acciones.

### **3.2 Redes Convolucionales (ConvNets):**

Son redes neuronales que se utilizan para **procesar imágenes, audio y texto.** Utilizan una técnica llamada **"convolución" para identificar patrones y características en los datos de entrada.** Son recomendadas para problemas de visión por computadora, como la detección de objetos o la clasificación de imágenes.

### **3.3 Redes Recurrentes (RNNs):** 

Son redes neuronales que utilizan conexiones cíclicas para **mantener una memoria a corto plazo de los patrones vistos en los datos de entrada.** Son recomendadas para problemas de secuencias, como la **generación de texto o la predicción de series temporales.**

### **3.4 Redes Autoencoder:** 

Son redes neuronales que tienen como objetivo **aprender una representación compacta de los datos de entrada**, utilizando una parte codificadora y otra decodificadora. Estas redes están recomendadas para problemas de **reducción de dimensionalidad y detección de anomalías.**

### **3.5 Redes Generativas Adversarias (GANs):** 

Son redes neuronales que **consisten en dos partes: un generador y un discriminador.** El generador produce datos falsos, mientras que el discriminador intenta identificar si los datos son reales o falsos. Estas redes están recomendadas para **problemas de generación de datos, como la generación de imágenes o la creación de música.**

## **4. Las capas y sus parámetros.**

TensorFlow **permite el uso de diferentes tipos de capas para construir redes neuronales.** Algunas de las capas más comunes que se pueden encontrar en TensorFlow son:

1. **Dense layer:** Es una **capa completamente conectada**, donde cada neurona recibe una entrada de todas las salidas de la capa anterior y genera una salida para la capa siguiente. Es la capa más utilizada en las redes de tipo MLP.

2. **Convolutional layer:** Es una capa utilizada en redes neuronales de procesamiento de imágenes para **detectar patrones locales en una imagen**. Cada neurona de la capa de convolución recibe una entrada de una región específica de la capa anterior y produce una salida.

3. **Recurrent layer:** Es una capa utilizada en **redes neuronales de procesamiento de secuencias para modelar dependencias a lo largo del tiempo.** Cada neurona de una capa recurrente recibe tanto una entrada de la capa anterior como de la salida de la misma neurona en el paso temporal anterior.

4. **Dropout layer:** Es una capa utilizada para **prevenir el sobreajuste en redes neuronales.** Consiste en apagar aleatoriamente algunas neuronas durante el entrenamiento, lo que hace que la red sea más resistente a la entrada de datos no vistos.

5. **Pooling layer:** Es una capa utilizada en redes neuronales de procesamiento de imágenes para **reducir la dimensionalidad de la imagen**. Cada neurona de una capa de pooling recibe una entrada de una región específica de la capa anterior y produce una salida agregada.

6. **Flatten layer:** Se utiliza para **aplanar un tensor de forma N-dimensional a un tensor de forma unidimensional.**  La capa Flatten es útil en situaciones donde necesitas aplanar un tensor antes de conectarlo a una capa totalmente conectada en una red neuronal.

**En este tema nos vamos a centrar en el uso de capas Densamente Conectadas.**

#### **4.1.1 El Perceptrón**

* **Es una versión muy simplificada de una neurona.** 

* **La unidad básica de una red neuronal es el perceptrón.** La operación que realiza éste, es tan sencilla como una función lineal. 

* ¿Te acuerdas de la ecuación de la recta? Algo así como:
> **y = ax + b.** 

* Pues en el caso del perceptrón, **tenemos tantas variables x como datos de entrada**, y en vez de usar el coeficiente "a" multiplicando a "x", usamos un tipo de coeficientes que se llaman **"pesos" o "weights".** 

* A esta suma se le añade un término adicional llamado **sesgo o bias (b)**. 

* Con esa nomenclatura, tendríamos que la ecuación lineal de un perceptrón sería:

![Perceptron](https://drive.google.com/uc?id=1Hdf8QOMdmhK84E-d8gmkBTqocW-C6Qx7)

![Perceptron con detalle](https://drive.google.com/uc?id=1LTU7OxMyT38zrJTHnyb9iWiOU3QHIv81)

**Asi que tenemos:**

> ***Un perceptrón toma las entradas, x1 , x2 , …, xn , las multiplica por pesos, w1 , w2 , …, wn y agrega el término de sesgo, b, luego calcula la función lineal, z en la que se aplica una función de activación, f para obtener la salida, y .***

#### **4.1.2 Componentes Principales del Perceptrón**

El perceptrón está formado por una serie de componentes como:

* **Entrada.** 
  Las entradas en el algoritmo del perceptrón se entienden como **x1, x2, x3, x4 y así sucesivamente.**

* **Pesos.** 
  * Son valores que se planifican a lo largo de la sesión de preparación del perceptrón. Los pesos ofrecen un valor preliminar al inicio del aprendizaje del algoritmo. 
  * **Con la ocurrencia de cada inexactitud de entrenamiento, los valores de los pesos se actualizan.** Estos se representan principalmente como **w1, w2, w3, w4 y así sucesivamente.**

* **Sesgo o Bias**  
  También toma un valor numérico. Se suma a la suma ponderada de insumos. El propósito de incluir un término de sesgo es cambiar la función de activación de cada perceptrón para que no obtenga un valor cero. En otras palabras, si todas las entradas x1 , x2 , …, xn son 0, z es igual al valor del sesgo.

* **Función de activación.** 
  * Cada función de activación, toma un único número y **realiza una determinada operación matemática fija sobre él.** 
  * **Hay varias funciones de activación**, las más comunes son la Sigmoide o la ReLU. Las veremos con detalle más adelante.


* **Suma ponderada.** Es la suma de cada valor de entrada o característica asociada con el valor de peso correspondiente.


* **Salida.** **La suma ponderada se pasa a la función de activación** y cualquier valor que obtengamos después del cálculo es nuestra **salida predicha.**

#### **4.1.3 Modelo de perceptrón multicapa**

Un modelo de perceptrón multicapa tiene una estructura similar a la de un modelo de perceptrón de una sola capa con más número de capas ocultas. También se denomina **algoritmo de retropropagación.** 

Se ejecuta en dos etapas: 

1. **La etapa hacia adelante.**
2. **La etapa hacia atrás.**

![](https://drive.google.com/uc?id=1m0qNLFjb2ajcqcrvknbgxwyTK4nD328E)

---

* En la etapa hacia adelante, las funciones de activación se originan desde la capa de entrada a la capa de salida

* En la etapa hacia atrás, el error entre el valor real observado y el valor demandado se utiliza para **modificar los pesos y que la Red Neuronal Aprenda**

* En resumen, el perceptrón multicapa puede tratarse como una red de numerosas neuronas artificiales sobre capas variadas, **la función de activación ya no es lineal, en su lugar se despliegan funciones de activación no lineales.**

**En realidad esto es más sencillo de lo que creemos, lo hemos hecho en infinidad de problemas...**

### **4.2 Dense. Capa de Red Densamente Conectada o Fully Conected**

![Densa](https://drive.google.com/uc?id=1Xv-rpkohNQz2MywLev32RT9lo--SCc2t)

La capa o "layer" de tipo Dense, es la que representa verdaderamente una red neuronal con **todos los nodos conectados con las diferentes variables de entrada**.

En la imagen superior, hay tres neuronas o perceptrones que operan con combinaciones de las variables de entrada. A ésto, se le llama **red "densamente" conectada.**

La clase Dense, crea un objeto que implementa la operación **output = activation(dot(input, kernel) + bias),** donde:

* **Activation**
es la función de activación que actúa sobre el resultado que se da en cada neurona.

* **Kernel**
es la matriz de coeficientes o pesos "w" que se generan de forma aleatoria

* **Bias**
es el vector del sesgo que solo es distinto de cero si se pasa como argumento True.

Estos parámetros, entre otros, se pueden encontrar en el esquema:

[Documentacion Tensorflow](https://www.tensorflow.org/api_docs/python/tf/keras/layers/Dense)




```python
tf.keras.layers.Dense(
units,
activation=None,
use_bias=True,
kernel_initializer="glorot_uniform",
bias_initializer="zeros",
kernel_regularizer=None,
bias_regularizer=None,
activity_regularizer=None,
kernel_constraint=None,
bias_constraint=None,
**kwargs
)
```

**El parámetro "units"** se refiere al **número de neuronas que debe tener la capa.** <font color='green'>**Debe ser un número entero y positivo.**</font>

### **5.1 Entropía Cruzada**

En general, para los modelos basados en neuronas, es necesario utilizar lo que se conoce como cálculo de la **"entropía cruzada"**, en contraposición al error cuadrático medio o MSE que se venía utilizando en otros modelos.

Vale, pero qué es esto de **Entropía Cruzada... 😧😧😧**

>>La entropía cruzada **compara las predicciones del modelo con los resultados reales de los datos de entrenamiento y calcula cuánta información se pierde en el proceso**. En otras palabras, <font color='green'>**mide como de diferente son las predicciones del modelo de los resultados reales y cuánta incertidumbre hay en la predicción.**</font>

La **entropía cruzada se utiliza comúnmente como función de pérdida** en modelos basados en neuronas para el aprendizaje supervisado.

### **5.3 Descenso del Gradiente**

* El descenso de gradiente es un algoritmo utilizado en el aprendizaje automático y la optimización de redes neuronales que nos permite ajustar los pesos y sesgos (parámetros) de la red para minimizar la función de pérdida. 

* **¿¿Y esto para qué se utiliza??** 😰😰😰

* <font color='green'>**El descenso de gradiente**</font>. **nos ayuda a encontrar los parámetros que hacen que nuestra red neuronal se ajuste mejor a los datos de entrenamiento.**
 

* Se realiza a través de un **proceso iterativo.** 

En resumen, el descenso del gradiente es un algoritmo que se utiliza para buscar el mínimo **GLOBAL** de una función de coste y con ello <font color='green'>**CONSTRUIR MODELOS DE APRENDIZAJE AUTOMÁTICO MÁS EFICIENTES.**</font>




**Vamos a verlo con un ejemplo.**

>Imagina que eres un montañero que está perdido en en una montaña del pirineo, inmerso en una densa niebla que no te permite ver nada y que solo puedes sentir la pendiente del suelo debajo de tus pies.

Una buena estrategia para llegar al fondo del valle rápidamente es ir **cuesta abajo en dirección a la pendiente.**

* Para el primer paso, tienes 3 opciones:

  * **Seguir recto**
  * **A la izquierda** 
  * **A la derecha.**

* Después del primer metro, debes elegir nuevamente.

* Siempre quieres encontrar el mejor siguiente paso.

* Repites esto hasta llegar al fondo.

**El descenso de gradiente hace lo mismo.**

![Descenso del montañero](https://drive.google.com/uc?id=1LIY8NUnt1i0MJJVOaPwhmSXoECBlRx8s) 



#### **5.3.1 Tipos de Descenso del Gradiente**

El algoritmo cuenta con **varias versiones dependiendo del número de muestras que introduzcamos** a la red en cada iteración:

* **Descenso del gradiente estocástico (SGD Stochastic Gradient Descent):** 

  * Se introduce una <font color='green'>**única muestra aleatoria en cada iteración**.</font> El gradiente se calculará para esa muestra concreta, lo que supone la introducción de la deseada aleatoriedad, **dificultando así el estancamiento.** **El problema de esta versión es su lentitud,** ya que necesita de muchas más iteraciones, y además no aprovecha los recursos disponibles.
  * **Muy eficiente en grandes conjuntos de datos, pero también puede hacer que el algoritmo se atasque en óptimos locales.**


* **Descenso del gradiente en lotes (o Batch Gradient Descent):** 

  * <font color='green'>**Todos los datos disponibles se introducen de una vez.**</font> Esto **supondrá problemas de estancamiento**, ya que el gradiente se calculará usando siempre todas las muestras, y llegará un momento en que las variaciones serán mínimas. 
  * Como regla general: siempre nos conviene que **la entrada a una red neuronal tenga algo de aleatoriedad.**
  Solo recomendado para algunas tecnicas que requieren muchos puntos en lugar de 1.
  * Este método es especialmente útil para **evitar atascarse en los mínimos locales.**


* **Descenso del gradiente (estocástico) en mini-lotes (o Mini-Batch Gradient Descent):** 

  * En lugar de alimentar la red con una única muestra, **se introducen N muestras en cada iteración;** conservando las ventajas de la primera versión y consiguiendo además que el entrenamiento sea más rápido debido a la paralelización de las operaciones. 
  * Nos quedamos pues con esta modificación del algoritmo, eligiendo un valor de N que nos aporte un buen balance entre aleatoriedad y tiempo de entrenamiento.

## **6. Optimizadores**

Un optimizador es un **algoritmo específico que implementa el descenso del gradiente** (u otro método de optimización) para **actualizar los parámetros del modelo** y minimizar la función de pérdida.

El optimizador es otro de los argumentos que se requieren en el entrenamiento del modelo.

---

**Tipos de Optimizadores en Keras**

 Keras ofrece varios optimizadores para entrenar modelos de redes neuronales. Algunos de los optimizadores más comunes disponibles en Keras son:

* **SGD:** Descenso del gradiente estocástico. Es un optimizador básico que ajusta los parámetros del modelo en la dirección opuesta al gradiente de la función de pérdida con respecto a los parámetros.

* **RMSprop:** Es un optimizador que utiliza una tasa de aprendizaje adaptativa para ajustar los parámetros del modelo. La tasa de aprendizaje se ajusta dividiendo el gradiente por una media móvil de la raíz cuadrada de los gradientes recientes.

* **Adagrad:** Es un optimizador que también utiliza una tasa de aprendizaje adaptativa. La tasa de aprendizaje se ajusta dividiendo el gradiente por la raíz cuadrada de la suma acumulada de los cuadrados de los gradientes recientes.

* **Adadelta:** Es una extensión del optimizador Adagrad que busca reducir su agresiva disminución de la tasa de aprendizaje.

* **Adam:** Es un optimizador que combina las ideas de RMSprop y Adagrad. Utiliza estimaciones adaptativas del momento del primer y segundo orden para ajustar los parámetros del modelo.

Todos estos optimizadores **tienen diferentes hiperparámetros que puedes ajustar** para mejorar el rendimiento en tu problema específico.

Se puede encontrar más información sobre cada uno de ellos en la [documentación de Keras](https://keras.io/api/optimizers/)

En definitiva, los optimizadores se basan en **favorecer esa evolución hacia el mínimo:**

## **6.1 Ratio de Aprendizaje o "Learning Rate**

**Ratio de aprendizaje o "Learning Rate"**
es el <font color='green'>**parámetro que controla la magnitud con la que modificamos los pesos en función de la pendiente de la función de coste**.</font> 

* Podríamos decir que es un **factor que amplifica el efecto de la pendiente** de la función de coste. 

* **Si el learning rate es pequeño**, aunque estemos en una zona de fuerte pendiente, **se avanzará a pasos pequeños**, y en la zona de baja pendiente, según vamos alcanzando el mínimo, cada vez avanzará mucho más despacio, lo que <font color='green'>**hará el entrenamiento muy lento.**</font> 

* **Si el learning rate**, por el contrario, es **demasiado alto**, en la zona de pendiente muy elevada, hará pegar un salto al factor de corrección de los pesos, que nos iremos a la otra cara de la función de coste. <font color='green'>**El proceso de entrenamiento será muy inestable e incluso podría no converger.**</font>

* **El rango de un learning rate de partida está entre 0,0001 y 0,001**. A veces hay que tomar uno más alto o más bajo de los valores de ese rango, pero podemos empezar por ahí e ir probando.

![Learning rates](https://drive.google.com/uc?id=1U2oue3BuROz6xyJ2mm1LIBvWx_5BCul0)


**En keras, al llamar a la función de optimización, podemos pasarle un valor de learning rate como argumento por clave:**


```python
opt = keras.optimizers.Adam(learning_rate=0.001)

model.compile(loss='categorical_crossentropy', optimizer=opt)
```

* Por ejemplo, el optimizador SGD tiene hiperparámetros como la tasa de aprendizaje (learning_rate), el impulso (momentum) y el impulso de Nesterov (nesterov).

Para ajustar los parámetros del optimizador en Keras, puedes crear una instancia del optimizador y pasar los valores de los hiperparámetros como argumentos al constructor. Luego, puedes pasar la instancia del optimizador al método compile del modelo. Aquí hay un ejemplo de cómo ajustar la tasa de aprendizaje y el impulso del optimizador SGD en Keras:

```python
# Crear una instancia del optimizador SGD con una tasa de aprendizaje y un impulso específicos
optimizer = keras.optimizers.SGD(learning_rate=0.01, momentum=0.9)

# Compilar el modelo con el optimizador personalizado
model.compile(loss='binary_crossentropy', optimizer=optimizer, metrics=['accuracy'])
```

## **8.Evaluación del Modelo**

Una vez hemos **creado la red neuronal, la hemos compilado y la hemos entrenado**, solo resta evaluar cómo de buenos son nuestros resultados.

Para evaluar el modelo, recurrimos a la función **evaluate** y guardamos los valores que nos devuelve, el **error o loss, y la precisión o accuracy.**

<font color='blue'>**El "loss" (o pérdida):**</font> 
  * Es una medida que se utiliza para evaluar **cómo de bien está prediciendo el modelo con los datos de entrenamiento.** 
  * **Mide la diferencia entre la salida del modelo y la salida esperada para un conjunto de datos de entrenamiento dado.** 
  * El objetivo del entrenamiento de una red neuronal es **minimizar la pérdida o loss.**

  * **Será un valor numérico. Cuanto más bajo sea, mejor.**

<font color='blue'>**La "accuracy" (precisión)**</font>
* Es otra medida importante para evaluar el rendimiento de una red neuronal. 

* **Mide la proporción de predicciones correctas que hace la red neuronal en el conjunto de datos de prueba.** 

* Es una medida que indica **cómo de bien** generaliza el modelo con **datos nuevos que nunca antes había visto.**

* **Será un ratio, cuanto más cerca de 1 mejor. **Puede traducirse a % multiplicandolo por 100.

---

Si estamos construyendo un modelo de clasificación de imágenes de perros y gatos, **la pérdida mide cuán diferentes son las salidas del modelo de las salidas esperadas** para el conjunto de entrenamiento, mientras que **la precisión mide cuántas imágenes de perros y gatos el modelo clasifica correctamente en el conjunto de prueba.**

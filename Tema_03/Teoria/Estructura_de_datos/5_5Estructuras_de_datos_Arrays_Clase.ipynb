## 1. Qué son los Arrays en Python


* Los Arrays en Python no son los Arrays de los lenguajes de programación convencionales como C y Java, sino más cercanos a las listas. Una lista puede ser una colección de elementos homogéneos o heterogéneos y puede contener enteros, cadenas u otras listas. 


* Se puede acceder a los elementos individuales a través de índices. Los Arrays de Python están indexadas a cero. Esto es igual a las estructuras que hemos estudiado e incluso en otros lenguajes como C o Java.


* Un Array es una estructura de datos que **almacena valores del mismo tipo de datos.** En Python, esta es la **principal diferencia entre arrays y listas.**


* Son **más rápidos y ocupan menos memoria que las listas.** 


* Mientras que las listas de python pueden contener valores correspondientes a diferentes tipos de datos, **los arrays en python solo pueden contener valores correspondientes al mismo tipo de datos.**


* Para usar Arrays, debemos **importar el módulo array.** Esto se debe a que los arrays no es un tipo de datos fundamentales como las cadenas, enteros, etc.

### 5. Extendiendo un Array usando Extend()

Los array de python se pueden ampliar con más de un valor utilizando el método extend().

```python
my_array = array('i', [1,2,3,4,5])
my_extnd_array = array('i', [7,8,9,10])
my_array.extend(my_extnd_array)

my_array#podemos ver que el array my_array ha sido ampliado con los valores de my_extnd_array
```

### Ejercicio 6 - Eliminando primera aparición

Escribe un programa en python que elimine la primera aparición de un elemento introducido por el usuario. Antes de eliminarlo comprueba si realmente existe en el array.

array1 = array('i', [1, 3, 5, 3, 7, 1, 9, 3])

valor a eliminar = 3

array2 = array('i', [1, 5, 3, 7, 1, 9, 3])

valor a eliminar = 1

array3 = array('i', [1, 5, 3, 7, 1, 9, 3])

valor a eliminar = 8

```python
from array import array

array1 = array('i', [1, 3, 5, 3, 7, 1, 9, 3])
array2 = array('i', [1, 5, 3, 7, 1, 9, 3])
array3 = array('i', [1, 5, 3, 7, 1, 9, 3])

def pedirValorEliminar():
    valorEliminar = int(input("Introduce que valor vas a eliminar del array: "))
    return valorEliminar

def eliminarValor(array, valorEliminar):
    if valorEliminar in array:
        for elemento in array:
            if elemento == valorEliminar:
                array.remove(elemento)
    else:
        print(f'El numero {valorEliminar} no se encuentra en el array.') 
        
eliminarValor(array1, pedirValorEliminar())
print(f'Array 1 postborrado: {array1}')
eliminarValor(array2, pedirValorEliminar())
print(f'Array 2 postborrado: {array2}')
eliminarValor(array3, pedirValorEliminar())
print(f'Array 3 postborrado: {array3}')
```

### Ejercicio 7. ¿Hay duplicados?

Escribe un programa de Python para encontrar si un array de enteros contiene algún elemento duplicado. Devuelve verdadero si algún valor aparece al menos dos veces en dicha matriz y devuelve falso si cada elemento es distinto. 

Ejemplos de salida:

array1 = array('i', [1, 5, 3, 7, 1, 9, 3])

Verdadero

array2 = array('i', [3, 7, 1, 9, 3])

Verdadero

array3 = array('i', [5, 3, 7, 1, 9])

Falso

```python
from array import array

array1 = array('i', [1, 5, 3, 7, 1, 9, 3])
array2 = array('i', [3, 7, 1, 9, 3])
array3 = array('i', [5, 3, 7, 1, 9])

def hayDuplicados(array):
    for elemento in array:
        if array.count(elemento) > 1:
            return True
    return False

print(f'Hay duplicados en el Array 1?: {hayDuplicados(array1)}')
print(f'Hay duplicados en el Array 2?: {hayDuplicados(array2)}')
print(f'Hay duplicados en el Array 3?: {hayDuplicados(array3)}')
```

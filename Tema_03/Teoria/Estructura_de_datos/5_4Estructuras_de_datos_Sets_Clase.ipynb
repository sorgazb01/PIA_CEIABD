## 1. Qué son los Sets o Conjutos

* **Los sets (o conjuntos)** son una colección de elementos **no ordenados y sin duplicados.** Son útiles cuando necesitas almacenar datos únicos y realizar operaciones matemáticas de conjuntos como unión, intersección y diferencia.

### 1.1 Características de los Conjuntos en Python:

1. **No tienen elementos duplicados:** Si añades un elemento que ya existe en el conjunto, no se añadirá de nuevo.
2. **Son no ordenados:** No mantienen el orden de los elementos, a diferencia de las listas.
3. **Los elementos deben ser inmutables:** Por ejemplo, puedes usar números, cadenas o tuplas, pero no listas o diccionarios.
4. **Son mutables:** Puedes añadir o eliminar elementos del conjunto (excepto en un caso especial).


### 1.2 Tipos de Conjuntos en Python
Hay dos tipos de conjuntos integrados en el lenguaje Python:

1. **Conjunto estándar (set)**
    * Es el tipo más común.
    * Se define utilizando llaves {} o la función set().
2. **Frozenset - Que son inmutables y hashables**
    * Este tipo de conjunto no se puede modificar una vez creado.
    * Se utiliza cuando necesitas un conjunto que sea constante.
    * Se crea usando la función frozenset().
    
Nosotros nos centraremos en los **Sets**

### 2. Creación de un Set

Crear un conjunto es bastante sencillo. Podemos crearlos **agregando una serie de caracteres separados por comas dentro de llaves (igual que en los diccionarios)** o podemos pasar una secuencia a la función integrada set().

Veamos un ejemplo:

```python
my_set = {"a", "b", "c", "c"}
print(my_set)
type(my_set)
```

Ahora vamos a crear un conjunto usando la función **Set()**

```python
mi_lista = [1, 2, 3, 4]

my_set = set(mi_lista)# hemos creado un set a partir de una lista.
#mostrar resultado:
```

```python
type(my_set)
```

```python
my_set
```

```python
# Intentamos crear un conjunto con un elemento mutable (una lista)
try:
    conjunto = {1, 2, [3, 4]}  # La lista [3, 4] es mutable
    print(conjunto)
except TypeError as e:
    print(f"Error: {e}")
```

```python
# Intentamos crear un conjunto con un elemento mutable (una lista)
try:
    dic ={"clave1":1}
    conjunto = {1, 2, dic}  # La lista [3, 4] es mutable
    print(conjunto)
except TypeError as e:
    print(f"Error: {e}")
```

**<u>Explicación</u>**

Esto sucede porque los conjuntos están basados en una **estructura de hash** para garantizar la unicidad de los elementos. Las listas, al igual que los diccionarios, al ser mutables, no tienen un valor de hash fijo, lo que los hace **incompatibles con conjuntos.**

Si necesitas algo similar a una lista como elemento, puedes usar una tupla (que es inmutable y compatible con conjuntos):

```python
# Usamos una tupla en lugar de una lista
conjunto = {1, 2, (3, 4)}
print(conjunto)  # Salida: {1, 2, (3, 4)}
```

### 3. Acceso a elementos del Set

Podemos verificar si un elemento está en un conjunto utilizando el **operador in de Python.**

```python
my_set = {"a", "b", "c", "d"}
"a" in my_set
```

#### Importante:

* Los conjuntos **no nos permiten usar el corte o similares** (slicing) para acceder a miembros individuales del conjunto. 
* En cambio, **podemos iterar sobre un conjunto.** Podemos hacerlo usando un bucle while o un bucle for.

```python
for item in my_set:
    print(item)
```

#### Importante:

* **Podemos acceder a elementos en conjuntos mucho más rápido que las listas.**
* Una lista de Python iterará sobre cada elemento hasta que encuentre el elemento que está buscando. 
* Cuando buscamos un elemento en un conjunto, actúa de manera muy similar a un diccionario y **lo encontraremos de inmediato o no lo encontraremos en absoluto.**

### 6. Eliminación de elementos

Podemos eliminar items de un set de diferentes formas. Podemos usar:

* **remove()**

* **discard()**

* **pop()**


#### Usando .remove()

El metodo intentará eliminar el elemento especificado del set.

```python
my_set = {"a", "b", "c", "c"}
my_set.remove('a')
print(my_set)
```

Si intentamos borrar un elemento que no está en el set, devolverá un error:

```python
my_set = {"a", "b", "c", "c"}
my_set.remove('f')
```

#### Usando .discard()

Funciona casi exactamente de la misma manera que remove() en el sentido de que eliminará el elemento especificado del conjunto. 

**La diferencia con remove es que no lanzará un error si intengamos borrar un elemento que no está en el set.**

```python
my_set = {"a", "b", "c", "c"}
my_set.discard('f')
```

#### Nota:

Si deseas poder detectar un error cuando intentas eliminar un elemento que no existe, usa remove(). Si eso no te importa, entonces discard() podría ser una mejor opción.

#### Usando Pop()

Borra y devuelve un elemento **aletorio** del set

```python
my_set = {"a", "b", "c", "c", "d","f"}
print(my_set)
```

```python

my_set.pop()
print(my_set)
```

```python
my_set.pop()
print(my_set)
```

#### Nota:

Si el set está vacio y tratamos de borrar un elemento con pop(), devolverá un error.

### 8. Operaciones con Sets

Los conjuntos nos proporcionan algunas operaciones comunes como:

* **union()** 

* **intersection()** 

* **difference()** 

Vamos a ver con detalle cada una de estas peraciones

#### Union()

Combina dos sets y retorna un nuevo set con los datos de los dos conjuntos.

```python
first_set = {'one', 'two', 'three'}
second_set = {'orange', 'banana', 'peach'}
print(f"Primer set\n {first_set}")

print(f"Set con la union:\n{first_set.union(second_set)}")
```

#### Intersection()
Retorna un nuevo set con los elementos que son comunes entre los dos.

```python
first_set = {'one', 'two', 'three'}
second_set = {'orange', 'banana', 'peach', 'one'}
first_set.intersection(second_set)
```

El resultado es un set que unicamente tiene un elemento "one" que es único común a los dos sets.

#### Difference()

Resulta en otro conjunto con los valores del primer conjunto menos los valores del segundo conjunto.

```python
A = {'one', 'two', 'three'}
B = {'three', 'four', 'one'}
C = A.difference(B)
C
```

```python
A = {'one', 'two', 'three'}
B = {'three', 'four', 'one'}
C = A - B
C
```

#### Nota:

Cuando llamamos a difference con el primer set, nos retorna un set con el elemento "two" es la única cadena que no se encuenta en el segundo set.

**Es equivalente a hacer C= A - B**

```python
C = A - B

C
```

```python
C = B - A
D = B.difference(A)

print(f"Conjunto C --> B - A: {C}")
print(f"Conjunto D --> B.Difference(A): {D}")
```

### Ejercicio 4.

Dadas una cadena verifique si es **pangrama** o no.

Una cadena es pangrama cuando contiene todas las letras del alfabeto.

a) Desarrolla un algoritmo que nos diga si un texto es pangrama.


* **Entrada:**

    * Un jugoso zumo de piña y kiwi bien frío es exquisito y no lleva alcohol.
         
    * **Resultado:**
    
        * True, la cadena es un pangrama

#### Solución:

### Versión 1 restando al alfabeto nuestra cadena, si el resultado es el set vacio, es un pangrama.

```python
from string import ascii_lowercase as asc_lower

def esPangrama(texto):
    letrasTotales = set(asc_lower)
    letrasTexto = set(texto.lower())
    letrasFaltantes = letrasTotales - letrasTexto
    if(len(letrasFaltantes) == 0):
        return True
    else:
        return False
    
texto = 'Un jugoso zumo de piña y kiwi bien frío es exquisito y no lleva alcohol.'
resultado = esPangrama(texto)
print(f'Pangrama: {resultado}')
```

### Versión 2, contando los caracteres del alfabeto y los del set de nuestra cadena, si son iguales es pangrama

```python
from string import ascii_lowercase as asc_lower

def esPangrama(texto):
    letrasTotales = set(asc_lower)
    letrasTexto = set()
    textoMinusculas = texto.lower()
    for letra in textoMinusculas:
        if letra.isalpha():
            letrasTexto.add(letra)

    letrasComunes = letrasTotales & letrasTexto

    if len(letrasTotales) == len(letrasComunes):
        return True
    else:
        return False
texto = 'Un jugoso zumo de piña y kiwi bien frío es exquisito y no lleva alcohol.'
resultado = esPangrama(texto)
print(f'Pangrama: {resultado}')
```

### Ejercicio 5.

Se nos dan una lista con n listas (matriz de listas) de cualquier tamaño que pueden tener elementos comunes.

Necesitamos combinar todas estas matrices de tal manera que cada elemento deba aparecer solo una vez y los elementos deben estar ordenados.


* **Entrada:**

    matriz = [
    
    [1, 2, 2, 4, 3, 6],

    [5, 1, 3, 4],

    [9, 5, 7, 1],

    [2, 4, 1, 3]
    
    ]

* **Salida:**

    [1, 2, 3, 4, 5, 6, 7, 9]

#### Solución:



```python
def combinar_listas(matriz):
    elementos_unicos = set()
    
    for lista in matriz:
        for elemento in lista:
            elementos_unicos.add(elemento)
    
    resultado = sorted(elementos_unicos)
    return resultado

matriz = [
    [1, 2, 2, 4, 3, 6],
    [5, 1, 3, 4],
    [9, 5, 7, 1],
    [2, 4, 1, 3]
]

resultado = combinar_listas(matriz)
print(resultado)
```

## Resumen

* Los sets o conjunto son tipo de datos que se definen como una "colección desordenada de distintos objetos hashable (inmutables)".

* Podemos usar un conjunto para **pruebas de existencia, eliminar duplicados de una secuencia y calcular operaciones matemáticas, como intersección, unión, diferencia.**

* No registran la posición de los elementos ni el orden de inserción.

* No son compatibles con la indexación, el corte u otros comportamientos similares a secuencias que hemos visto con listas y tuplas.

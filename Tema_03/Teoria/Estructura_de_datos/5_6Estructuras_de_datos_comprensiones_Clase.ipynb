## 2. Comprensiones de Listas

* La comprensión de una lista nos permite **crear una lista a partir de otra colección,** como otra lista o diccionario. 

* Una lista por comprensión, como una lista, **comienza y termina con corchetes, []**

* También tiene un bucle for incorporado, todo en una sola línea. 

* Opcionalmente, esto puede ir seguido de cero o más cláusulas for o if (dentro de los corchetes).


---
### 2.1 Ejemplos de compresiones de listas sin Filtros.

```python
sequence = [1, 2, 3]
new_list = [x*2 for x in sequence]

print(new_list)
```

#### El anterior código es equivalente a:

```python
sequence = [1, 2, 3]
new_list = []
for x in sequence:
    new_list.append(x*2)
print(new_list)
    
```

Por lo general, cuando se usa una lista de comprensión, deseamos **hacer algo con cada uno de los elementos de la colección.** 

Vamos a ver algunos ejemplos:

* **Cuadrado de los 10 primeros números:**

```python
'''
En este ejemplo hace el cuadrado 10 primeros números y los almacenamos en una nueva lista,
todo en una única línea.
'''
new_list = [x ** 2 for x in range(10)]
new_list
```

* **Poner en mayúsculas una lista de palabras:**

```python
palabras = "Esto es Esparta!".split()
mayusculas = [palabra.upper() for palabra in palabras] 
mayusculas
```

* **Encontrar la longitud de cada palabra de una frase:**

```python
palabras = "Ser o no ser, esa es la cuestión.".split()
longitudes = [len(palabra) for palabra in palabras]
print(longitudes)
```

### 2.2 Comprensiones de listas con Filtros

* **Podemos agregar una instrucción if a una lista de comprensión como un tipo de filtro.**  

    * Supongamos que deseamos crear una lista que contenga solo los números impares en ese rango.  
    
    * Podemos usar una lista de comprensión con una declaración if para hacer eso en 1 línea:

* **Primeros impares:**

```python
impares = [x for x in range(10) if x % 2 != 0]
impares
```

* **Cuadrados de los primeros pares:**

```python
[n ** 2 for n in range(10) if n % 2 == 0]
```

```python
# Es equivalente a:
[n ** 2 for n in range(0, 10, 2)]#es mejor opción porque usa menos datos iniciales
```

* **Poner en mayúsculas aquellas palabras de una frase que están en minúsculas:**

```python
words = "Esto es Esparta!".split()
[word.upper() for word in words if word == word.lower()]

#el resultado solo contiene "ES" porque es la única palabra que está totalmente en minuscula
```

* **Encuentra la longitud de cada palabra que no tiene puntuación al lado:**

```python
palabras = "Ser, o no ser, esa es la cuestión.".split()
print(palabras)
longitudes = [len(palabra) for palabra in palabras if palabra.isalpha()]
print(longitudes)
```

### 2.2.1 Declaraciones condicionales al principio de la expresión.

* Las declaraciones condicionales se pueden colocar al final de la declaración para completar el filtrado simple, **pero, ¿qué sucede si desea cambiar el valor del miembro en lugar de filtrarlo?** 


* **En este caso, es útil poner la condición al principio de la expresión.**

**Observa su expresión:**

#### new_list = [expression (if conditional else ) for member in iterable]

```python
lista = [1, 5, 6, 7, 8,5, 9, 11, 3, 9]

#vamos a cambiar los impares por el valor 8
mis_pares = [x if (x % 2 == 0) else 8 for x in lista]
mis_pares
```

```python
lista = [1, 5, 6, 7, 8,5, 9, 11, 3, 9]

#vamos a cambiar los pares por 0 y los impares por el valor -1
mis_pares = [0 if (x % 2 == 0) else -1 for x in lista]
mis_pares
```

### Ejercicio 2 - El buscador de Aes.

Dada una frase cualquiera, transforma cada palabra aplicando el siguiente criterio:

* Si la palabra contiene **al menos una a la ponemos en mayuscula.**

* Si la palabra **NO contiene ninguna a irá con formato de título.**

a) Resuelve el problema sin comprensiones.

b) Soluciona el ejercicio usando comprensión.

### Solución:

```python
#a)
frase = 'En un lugar de La Mancha de cuyo nombre no quiero acordarme'

def convertirFrase(frase):
    nuevaFrase = ''
    for palabra in frase.split():
        if 'a' in palabra or 'A' in palabra:
            nuevaFrase += palabra.upper() + ' '
        else:
            nuevaFrase += palabra.capitalize() + ' '
    return nuevaFrase.strip()

print(convertirFrase(frase))
```

```python
#b)
def convertirFraseCompresion(frase):
    return ' '.join([palabra.upper() if 'a' in palabra.lower() else palabra.capitalize() for palabra in frase.split()])

print(convertirFraseCompresion(frase))
```

### Ejercicio 4 - Divisores de X

Crea un programa que pida al usuario un número comprendido entre el 1 el 300 y genere una lista con todos los valores divisores.

Resuelvelo sin y con comprensiones.

```python
#a) Sin comprensiones
def pedirNumero1a300():
    while True:
        numero = int(input('Introduce un numero entre 1 y 300: '))
        if 1 <= numero <= 300:
            return numero
        else:
            print('Error. Numero fuera del rango. ')
            
numero = pedirNumero1a300()

def obtenerDivisores(numero):
    divisores = []
    for i in range(1, numero + 1):
        if numero % i == 0:
            divisores.append(i)
    return divisores
        
print(f'La funcion sin compresion, devuelve los siguientes divisores para {numero}: {obtenerDivisores(numero)}')
```

```python
#b) Con comprensiones
def obtenerDivisoresCompresion(numero):
    return [i for i in range(1, numero + 1) if numero % i == 0]

print(f'La funcion con compresion, devuelve los siguientes divisores para {numero}: {obtenerDivisoresCompresion(numero)}')
```

### 2.3 Equivalencia con Bucle For

* Ahora que hemos visto algunas comprensiones de listas, es hora de crear una analogía con bucles for. 


* Si conoces los bucles for lo suficientemente bien, y si estudiamos esta equivalencia, dominaremos la compresión de listas.


* Una lista por comprensión **es equivalente a un ciclo for que llama consecutivamente al método .append de una (inicialmente) lista vacía:**


![Equivalencia comprension de lista con bucle for](equivalenca_comprension_lista.PNG)

![Equivalencia comprension de lista con bucle for](https://drive.google.com/uc?id=1EBHNvDqkFaKrLeSuEv2Hf17WazWLxUcw)

### Importante:

* Este es, de hecho, uno de los patrones más comunes para los que son útiles las listas de comprensión.


* Si encuentras un fragmento de código que inicializa una lista vacía y luego usa un bucle for para llenarlo con datos, **probablemente sea un buen caso de uso para la comprensión de una lista.**


* Por supuesto, esto no siempre se puede hacer de una manera sensata, **las listas de comprensión no están destinadas a reemplazar todos los bucles for**. Pero si tiene un ciclo corto que muestra la estructura anterior, probablemente podría reemplazarse por una lista de comprensión.


* **Te reto a que hagas precisamente eso. Revisa algún código tuyo y busca ese patrón. Luego, intenta reemplazarlo con una lista de comprensión.**


* Vamos a ver las comprensiones de lista de antes, con el equivalente de bucles:

* **Cuadrados de los pares:**

```python
cuadrados_pares = [n ** 2 for n in range(10) if n % 2 == 0]
```

```python
#equivalencia con bucle for e if:
cuadrados_pares = []
for n in range(10):
    if n % 2 == 0:
        cuadrados_pares.append(n ** 2)    
```

### 2.5 Anidamiento de Bucles

**Al igual que en los bucles for estándar, las listas por comprensión también pueden tener bucles anidados**

Vamos a verlo con una imagen:

![Anidamiento de bucles](anidamiento_bucles_comprensiones.PNG)

![Anidamiento de bucles](https://drive.google.com/uc?id=1VdxdZIWZWCJCG4j6RNo8yOiWJBJDdJ1U)

#### Un ejemplo de uso principal de esto es aplanar una lista de listas:

```python
lists = [[1, 2, 3], [4, 5, 6, 7], [8, 9]]
#sin comprensiones de listas como sería?
lista = []
for sublist in lists:
    for l in sublist:
        lista.append(l)
lista
        
```

```python
lists = [[1, 2, 3], [4, 5, 6, 7], [8, 9]]
#ahora con comprensiones, te atreves?
lista = [l for sublist in lists for l in sublist]
lista
```

### Importante:

* El segundo ciclo no necesita depender explícitamente del primero; puede iterar sobre otro iterable, para crear otra variable. 

* Cuando lo hagas, **todas las variables temporales que están pasando por iterables estarán disponibles para usarse a la izquierda.**

![Disposición de elementos en adinamiento de bucles por compresion](anidamiento_bucles_comprensiones2.PNG)

![Disposición de elementos en adinamiento de bucles por compresion](https://drive.google.com/uc?id=1uQ86IR0PkLvlDtbBWHg6-pZ3VEUfbfOj)

#### Este patrón surge naturalmente cuando deseamos combinar información de dos o más fuentes de datos:

```python
colores = ["rojo", "verde", "azul"]
prendas = ["pantalón", "camiseta"]
#como lo combinamos??
```

```python
[f"{prenda} {color}" for color in colores for prenda in prendas]
```

### Nota:

* Tenemos que tener en cuenta, aquí, que "combinar" la información significa **crear todos los diferentes emparejamientos con los datos de uno y el otro iterables.**


* **Si deseamos crear emparejamientos atravesando dos iterables en paralelo, entonces debemos usar zip y recordar que las listas han de tener el mismo tamaño**

```python
[f"{prenda} {color}" for color, prenda in zip(colores,prendas)]
```

```python
colores = ["rojo", "verde", "azul"]
prendas = ["pantalón", "camiseta", "falda"]
```

```python
[f"{prenda} {color}" for color, prenda in zip(colores,prendas)]
```

### 5. Cuándo NO usar comprensión en Python ☠️ ☣️

La comprensión de la lista es muy útil y puede ayudarnos a escribir un código bonito, legible, que sea fácil de leer y depurar, **pero no es una opción en todos los casos. Porque puede hacer que nuestro código se ejecute más lento o consuma más memoria.**

* Aunque la generación de listas es conveniente, **está limitada por la memoria y la capacidad de la lista es definitivamente limitada.** 


* Además, cuando la lista es grande, solo se accede a los primeros elementos, lo que también **provoca un desperdicio de memoria.**


* Entonces, si los elementos de la lista se pueden calcular de acuerdo con un cierto algoritmo, **¿se pueden calcular los elementos subsiguientes continuamente en el ciclo?** Esto elimina la necesidad de crear una lista completa, lo que ahorra mucho espacio.


#### En Python, este mecanismo de generar cálculos una y otra vez se llama generador. 



___
### 5.1 Comprensión de Generadores

La comprensión de generadores es una característica de Python que nos permite crear un generador de una manera más concisa y elegante. La sintaxis de la comprensión de generadores se parece a la de las listas por comprensión, pero en lugar de usar corchetes [], **utilizamos paréntesis ().**

**Se trata de colecciones cuyos elementos se crean a medida que se recorren en un bucle «for»** (o bien a medida que son pasados como argumento a la función incorporada next()). 

Por ejemplo, si queremos crear una lista de números pares del 1 al 100_000_000, podemos utilizar una lista por comprensión de la siguiente manera:



```python
%%timeit
pares = [i for i in range(1, 100_000_000) if i % 2 == 0]
```

**Para crear un generador en lugar de una lista, simplemente utilizamos paréntesis en lugar de corchetes:**

```python
%%timeit
```

```python

pares_generador = (i for i in range(1, 100_000_000) if i % 2 == 0)

```

```python
pares_generador
```

```python
#OJO si tenemos la función %%timeit cuando creamos el generador, la variable dnd almacenamos el generador, no va a ser
# reconocida fuera de esa celda.Para usar next tendriamos que quitar %%timeit que unicamente vale para medir tiempos.

# Obtenemos el primer valor del generador
print(next(pares_generador)) # imprime 2

# Obtenemos el segundo valor del generador
#print(next(pares)) # imprime 4

# Obtenemos el tercer valor del generador
#print(next(pares)) # imprime 6
```

* Esto crea un objeto que **genera progresivamente (a medida que son utilizados)** pares del 1 al 100_000_000, pero que se ejecuta casi instantáneamente porque los elementos no son creados hasta que no sean obtenidos con **next()**

```python
impares = (i for i in range(10) if i%2 !=0 )
```

```python
help(impares)
```

```python
try:
    valor = next(impares)
    print(valor)
except StopIteration:
    print("El generador ha llegado a su fin.")
```

Dado que los paréntesis se usan para esa situación en particular (en lugar de los corchetes o las llaves), si queremos generar una tupla usando el método de comprensión lo debemos indicar específicamente así:

```python
doubles = tuple(n * 2 for n in range(1, 11))
doubles
```

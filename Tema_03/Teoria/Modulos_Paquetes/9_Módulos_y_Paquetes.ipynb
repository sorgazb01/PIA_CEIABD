### 2.1 \___name___


A veces, se escribe un módulo (un archivo .py) que se puede ejecutar directamente, pero que, alternativamente, también se puede importar y reutilizar sus funciones, clases, métodos, etc, en otro módulo.

Todo módulo en python tiene un atributo especial llamado \__**name**__	que define el espacio de nombres en el que se está ejecutando. **Es usado para identificar de forma única un módulo en el sistema de importaciones.**

Por otro lado \__**main**__ es el nombre del ámbito en el que se ejecuta el código de nivel superior en el programa principal.


* **<div class="alert alert-success">Si el módulo se está ejecutando como programa principal, el intérprete pasa el valor \__main__  al atributo \__name__</div>**

* **<div class="alert alert-info">Si el módulo no es llamado como programa principal, sino que es importado desde otro módulo, el atributo __name__ pasa a contener el nombre del archivo.</div>**


Vamos a ver esto con un ejemplo.

```python

def suma(a,b):
    """Funcion que realiza la suma de a + b"""
    print("el resultado de la suma es: ",a+b)

def resta(a,b):
    """Funcion que realiza la resta de a - b"""
    print("el resultado de la resta es: ",a-b)

def multiplica(a,b):
    """Funcion que realiza la miltiplicacion de a * b"""
    print("el resultado de la multiplicación es: ",a*b)

def divide(a,b):
    """Funcion que realiza la division de a / b"""
    print("el resultado de la división es: ",a/b)

if __name__ == "__main__":
    print('Ejecutando como programa principal')
    suma(6,5)
```

* Básicamente, lo que hacemos usando **if  \__name__	== “\__main__”: es comprobar si el módulo ha sido importado o no.** 


* **Si no se ha importado (se ha ejecutado como programa principal) ejecuta el código dentro del condicional.**

En Resumen:

**if \__name__ == "\__main__":** 

<mark>**Es una convención establecida en python para saber si un módulo se ejecuta como programa principal o desde otro módulo. Es bueno que te acostumbres a utilizarlo en todos tus módulos**</mark>

**Módulo para validación de nombres de usuarios - usuarios.py**

Este módulo, deberá cumplir con los siguientes criterios de aceptación:

1.	El nombre de usuario debe contener un **mínimo de 6 caracteres y un máximo de 12.**


2.	El nombre de usuario debe ser **alfanumérico.**


3.	Nombre de usuario con **menos de 6 caracteres,** retorna el mensaje "El nombre de usuario debe contener al menos 6 caracteres".


4.	Nombre de usuario con **más de 12 caracteres,** retorna el mensaje "El nombre de usuario no puede contener más de 12 caracteres".

**Módulo para validación de contraseñas - contrasena.py.** 

Dicho módulo, deberá cumplir con los siguientes criterios de aceptación:

1.	La contraseña debe contener un **mínimo de 8 caracteres.**


2.	Una contraseña debe contener **letras minúsculas, mayúsculas, números y al menos 1 carácter no alfanumérico.**


3.	La contraseña **no puede contener espacios en blanco.**


4.	Contraseña válida, retorna "True".

#### usuario.py

```python
import string

def validar_nombre(nombre):
    """Función que se encarga de validar si el nombre recibido por prámetro es un nombre correcto.
       1 El nombre de usuario debe contener un mínimo de 6 caracteres y un máximo de 12.
       2 El nombre de usuario debe ser alfanumérico.
       3 Nombre de usuario con menos de 6 caracteres, retorna el mensaje "El nombre de usuario debe contener al menos 6 caracteres".
       4 Nombre de usuario con más de 12 caracteres, retorna el mensaje "El nombre de usuario no puede contener más de 12 caracteres".
       5 Nombre de usuario con caracteres distintos a los alfanuméricos, retorna el mensaje
       "El nombre de usuario puede contener solo letras y números".
       6 Nombre de usuario válido, retorna True.
    """
    #obtenemos todos los caracteres alfanumericos(minusculas y mayusculas)
    alfanumericos = string.ascii_letters + 'ñÑ' + string.digits
    if len(nombre) < 6:
        return (False,'El nombre de usuario debe contener al menos 6 caracteres')
    elif len(nombre) > 12:
        return (False,'El nombre de usuario no puede contener más de 12 caracteres')

    #comprobamos que cada caracter del nombre sea alfanumerico
    for c in nombre:
        if c not in alfanumericos:
            return (False,'El nombre de usuario puede contener solo letras y números')
    return (True,"")
```

#### contrasena.py

```python
"""
Modulo para validacion de contraseñas.

Dicho modulo, deberá cumplir con unos criterios de aceptación.

"""

__author__ = "Saul Lazaro Hermoso"
__copyright__ = "Python3"
__credits__ = ["Saul Lazaro Hermoso"]
__license__ = "GPL"
__version__ = "0.1"
__email__ = ""
__status__ = "Desarrollo"

import string

def validar_password(password):

    """ Esta función valida si la password cumple con los siguientes requisitos:
    1 La contraseña debe contener un mínimo de 8 caracteres.
    2 Una contraseña debe contener letras minúsculas, mayúsculas, números y al menos 1 carácter no alfanumérico.
    3 La contraseña no puede contener espacios en blanco.
    4 Contraseña válida, retorna "True".
    5 Contraseña no válida, retorna el mensaje "La contraseña elegida no es segura"""

    #letras, digitos y caracteres de puntuacion
    valores = string.ascii_letters + string.digits + string.punctuation + 'ñÑ'
    enc = True
    contador =0
    if len(password)>8:
        #comprobamos caracter por caracter
        for c in password:
            if c not in valores or c.isspace():
                enc = False
                break
            elif c in string.punctuation:
                #comprobamos si tiene al menos un caracter
                contador += 1
        #no ha encontrado caracteres no alfanumericos
        if contador == 0:
            enc = False
    else:
        enc = False
    #devolvemos el resultado final.
    if enc:
        return (enc,"")
    else:
        return (enc,"La contraseña elegida no es válida")
```

#### validador.py

```python
"""
Modulo que solicita al usuario el ingreso de un nombre de usuario y contraseña.

 Dicho modulo, valide utilizando los módulos generados en los dos ejercicios anteriores que el nombre de usuario y contraseña
 son correctos.

"""
__author__ = "Saul Lazaro Hermoso"
__copyright__ = "Python3"
__credits__ = ["Saul Lazaro Hermoso"]
__license__ = "GPL"
__version__ = "0.1"
__email__ = ""
__status__ = "Desarrollo"


import sys, usuario, contrasena

def main():
    nombre = input("Introduce un nombre de usuario: ")
    password = input("Introduce una contraseña: ")

    #comprobamos si el usuario cumple los requerimientos
    enc1,mensaje = usuario.validar_nombre(nombre)
    if enc1 == False:
        print(mensaje)
    else:
        print(mensaje)

    #comprobamos la fortaleza de la contraseña del usuario
    enc,mensaje = contrasena.validar_password(password)
    if enc == False:
        print(mensaje)
    else:
        print(mensaje)
    if enc1 and enc:
        print("Usuario y contraseña válidos")

if __name__ == '__main__':
    try:
        main()
    except KeyboardInterrupt:
        print("Saliendo...")
        sys.exit()
```

### 7. Paquetes

El concepto de paquete es una forma de ordenar el código dentro de Python, un paquete no es otra cosa que **una carpeta con un archivo \__init__.py** que le dicen a Python que es **un grupo de archivos(módulos) preparados para ser usados por algún programa.**

Cuando crees un paquete, **lo primero que debes hacer es crear el archivo \__init__.py dentro de la carpeta,** que será el que indica a Python que está dentro del contenido de un paquete para que acceda y lo use como tal. 

**Después de esto, ya puedes crear los módulos, script y subpaquetes necesarios.**

Ahora, si utilizas un script desde el mismo directorio donde se encuentra el paquete puedes acceder a los módulos, pero esta vez refiriéndote al paquete y al módulo, así que debes hacerlo con from ... import, utilizando cualquiera de las variantes que ya hemos visto:

**Importante:**

**`La jerarquía de paquetes es bastante sencilla, pues puedes crear subpaquetes dentro de paquetes con el único requerimiento de que se incluya un archivo __init__.py en cada uno de ellos.`**

#### strftime()

Puedes dar formato a fechas y horas de forma muy sencilla y potente para eso usa el método **strftime(format) que nos permite especificar como se muestra la fecha y hora:**

```python
from datetime import datetime
fecha = datetime.now()  # Calculo la fecha y hora actual
print(fecha.strftime("Las %H:%M:%S del %A %d de %B del %Y")) 
```

#### timedelta()

Para realizar operaciones y cálculos con fecha y horas, necesitas importar el método timedelta y crear una variable con **timedelta(days=0, seconds=0, microseconds=0, milliseconds=0, minutes=0, hours=0, weeks=0),** que usarás para operar con la fecha.

```python
from datetime import datetime, timedelta
fecha = datetime.now()
print(fecha.strftime("Las %H:%M:%S del %A %d de %B del %Y")) 
```

```python
una_semana = timedelta(weeks=1)
```

```python
una_semana_menos = fecha - una_semana
print(una_semana_menos.strftime("Las %H:%M:%S del %A %d de %B del %Y"))
```

```python
dos_semana_mas = fecha + una_semana + una_semana
print(dos_semana_mas.strftime("Las %H:%M:%S del %A %d de %B del %Y"))
```

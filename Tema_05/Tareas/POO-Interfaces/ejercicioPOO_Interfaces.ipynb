## Ejercicio.

1. Crea una interfaz para clases que representen adaptadores de base de datos.

    Los métodos comunes para el trabajo con Bases de Datos son:

    1. Conectar.
    2. Ejecutar consulta.
    3. Cerrar conexión.

2. Crea al menos 3 clases que utilicen la interfaz e implementen todos los métodos. No olvides crear los constructores de cada clase.

**Solución:**

```python
# Librerias interfaces
from abc import ABC, abstractmethod

# Interfaz Base de Datos
class AdaptadorBD(ABC):
    # Método abstracto para conectar con la base de datos
    @abstractmethod
    def conectar(self):
        pass
    # Método abstracto para ejecutar la consulta a la base de datos
    @abstractmethod
    def ejecutarConsulta(self):
        pass
    # Método abstracto para cerrar la conexión con la base de datos
    @abstractmethod
    def cerrarConexion(self):
        pass
```

```python
# Clase MySQL
class mySQL(AdaptadorBD):
    # Constructor, con los parámetros para conectarte a una base de datos.
    def __init__(self, host, usuario, contrasenia, baseDatos):
        self.host = host
        self.usuario = usuario
        self.contrasenia = contrasenia
        self.baseDatos = baseDatos
    # Método para conectar a la base de datos
    def conectar(self):
        print(f'MySQL: Conectando al sevidor ... {self.host}. Usuario: {self.usuario}, Base de datos: {self.baseDatos}, Contraseña: {self.contrasenia}')
    # Método para ejecutar la consulta a la base de datos
    def ejecutarConsulta(self, consulta):
        print(f'MySQL: Ejecutando consulta: {consulta}')
    # Método para cerrar la conexión con la base de datos
    def cerrarConexion(self):
        print('MySQL: Cerrando conexión')
```

```python
# Clase PostgresSQL
class postgreSQL(AdaptadorBD):
    # Constructor, con los parámetros para conectarte a una base de datos.
    def __init__(self, host, usuario, contrasenia, baseDatos):
        self.host = host
        self.usuario = usuario
        self.contrasenia = contrasenia
        self.baseDatos = baseDatos
    # Método para conectar a la base de datos
    def conectar(self):
        print(f'PostgreSQL: Conectando al sevidor ... {self.host}. Usuario: {self.usuario}, Base de datos: {self.baseDatos}, Contraseña: {self.contrasenia}')
    # Método para ejecutar la consulta a la base de datos
    def ejecutarConsulta(self, consulta):
        print(f'PostgreSQL: Ejecutando consulta: {consulta}')
    # Método para cerrar la conexión con la base de datos
    def cerrarConexion(self):
        print('PostgreSQL: Cerrando conexión')
```

```python
# Clase SQLite
class sqLite(AdaptadorBD):
    # Constructor, con los parámetros para conectarte a una base de datos.
    def __init__(self, host, usuario, contrasenia, baseDatos):
        self.host = host
        self.usuario = usuario
        self.contrasenia = contrasenia
        self.baseDatos = baseDatos
    # Método para conectar a la base de datos
    def conectar(self):
        print(f'SQLite: Conectando al sevidor ... {self.host}. Usuario: {self.usuario}, Base de datos: {self.baseDatos}, Contraseña: {self.contrasenia}')
    # Método para ejecutar la consulta a la base de datos
    def ejecutarConsulta(self, consulta):
        print(f'SQLite: Ejecutando consulta: {consulta}')
    # Método para cerrar la conexión con la base de datos
    def cerrarConexion(self):
        print('SQLite: Cerrando conexión')
```

```python
# Prueba clase MySQL
mysql = mySQL(host="localhost", usuario="root", contrasenia="1234", baseDatos="tienda")
mysql.conectar()
mysql.ejecutarConsulta("SELECT * FROM tienda WHERE precio < 50;")
mysql.cerrarConexion()

# Prueba clase PostgreSQL
postgres = postgreSQL(host="192.168.1.10", usuario="admin", contrasenia="password", baseDatos="pedidos")
postgres.conectar()
postgres.ejecutarConsulta("SELECT nombre, total FROM pedidos ORDER BY total DESC LIMIT 10;")
postgres.cerrarConexion()

# Prueba clase SQLite
sqlite = sqLite(host="192.168.1.10", usuario="admin", contrasenia="password", baseDatos="articulos")
sqlite.conectar()
sqlite.ejecutarConsulta("INSERT INTO articulos (nombre, cantidad) VALUES ('Monitor', 5);")
sqlite.cerrarConexion()
```

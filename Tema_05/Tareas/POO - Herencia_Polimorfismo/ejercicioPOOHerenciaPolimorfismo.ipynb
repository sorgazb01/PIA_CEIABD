**Implementa otra clase que se denomine `Gato` que es un mamífero y también un animal doméstico. Implementa su constructor y los siguientes métodos.**

* `maullar()`
Mostrará un mensaje del tipo: "Nombre dice: Miau Miuau Miau"

* `cazar()`

Mostrará un mensaje del tipo "El gato Nombre está cazando"

Para probar todo crea varas instancias de perro y gato y ejecuta los métodos:

* amamantar 

* dormir

* entrenar

* poner_vacunas

* comer

* ladrar/maullar

* jugar/cazar



```python
class Mamifero:
    def __init__(self, patas, pelaje):
        self.patas = patas
        self.pelaje = pelaje

    def amamantar(self):
        print('El mamifero está amamantando a sus crias')

    def dormir(self):
        print('El mamífero está durmiendo')


class AnimalDomestico:
    def __init__(self, dueno, nombre):
        self.dueno = dueno
        self.nombre = nombre

    def entrenar(self):
        print(f'El animal {self.nombre} está siendo entrenado por {self.dueno}')

    def poner_vacunas(self):
        print(f'Poniendo vacunas a {self.nombre}')

    def comer(self, alimento):
        print(f'El animal {self.nombre} está comiendo {alimento}')


class Perro(Mamifero, AnimalDomestico):
    def __init__(self, nombre, dueno, patas=4, pelaje='corto'):
        Mamifero.__init__(self, patas, pelaje)
        AnimalDomestico.__init__(self, dueno, nombre)

    def ladrar(self):
        print(f'{self.nombre} dice: Guau')

    def jugar(self):
        print(f'El perro {self.nombre} está jugando')


class Gato(Mamifero, AnimalDomestico):
    def __init__(self, nombre, dueno, patas=4, pelaje='suave'):
        Mamifero.__init__(self, patas, pelaje)
        AnimalDomestico.__init__(self, dueno, nombre)

    def maullar(self):
        print(f'{self.nombre} dice: Miau')

    def cazar(self):
        print(f'El gato {self.nombre} está cazando')
```

```python
# Creamos los objetos
perro1 = Perro('Bobi', 'Carlos')
perro2 = Perro('Rex', 'Ana', pelaje='largo')
gato1 = Gato('Michi', 'Laura')
gato2 = Gato('Pelusa', 'Pedro')

print('Perro 1: Bobi')
perro1.amamantar()
perro1.dormir()
perro1.entrenar()
perro1.poner_vacunas()
perro1.comer('pienso')
perro1.ladrar()
perro1.jugar()

print('Perro 2: Rex')
perro2.amamantar()
perro2.dormir()
perro2.entrenar()
perro2.poner_vacunas()
perro2.comer('hueso')
perro2.ladrar()
perro2.jugar()

print('Gato 1: Michi')
gato1.amamantar()
gato1.dormir()
gato1.entrenar()
gato1.poner_vacunas()
gato1.comer('atún')
gato1.maullar()
gato1.cazar()

print('Gato 2: Pelusa')
gato2.amamantar()
gato2.dormir()
gato2.entrenar()
gato2.poner_vacunas()
gato2.comer('croquetas')
gato2.maullar()
gato2.cazar()
```

### **Ejercicio 3 - 📹🎮 NetFlix 📹🎮**

Con los conocimientos adquiridos, desarrolla una aplicación que simule las altas/bajas de peliculas y socios de Netflix.
Nuestro Netflix tendrá dos atributos:
1. Lista de socios.
2. Lista de peliculas.

Será tarea tuya la creación de las clases socio y películas a tu gusto.

```python
class Pelicula:
    def __init__(self, titulo, genero, anio, duracion, director):
        self.titulo = titulo
        self.genero = genero
        self.anio = anio
        self.duracion = duracion
        self.director = director

    def mostrar(self):
        return (f"Título: {self.titulo} | Género: {self.genero} | Año: {self.anio} | Duración: {self.duracion} min | Director: {self.director}")

    def existePelicula(self, pelicula):
        if isinstance(pelicula, Pelicula):
            return self.titulo.lower() == pelicula.titulo.lower()
        return False

class Socio:
    def __init__(self, nombre, apellidos, email):
        self.nombre = nombre
        self.apellidos = apellidos
        self.email = email

    def mostrar(self):
        return (f"Nombre: {self.nombre} {self.apellidos} | Email: {self.email}")

    def existeSocio(self, socio):
        if isinstance(socio, Socio):
            return self.email.lower() == socio.email.lower()
        return False

class Netflix:
    def __init__(self):
        self.socios = []
        self.peliculas = []

    def darAltaSocio(self, socio):
        if socio in self.socios:
            print(f"El socio con email '{socio.email}' ya está registrado.")
        else:
            self.socios.append(socio)
            print(f"Socio '{socio.nombre} {socio.apellidos}' dado de alta correctamente.")

    def darBajaSocio(self, email):
        for socio in self.socios:
            if socio.email.lower() == email.lower():
                self.socios.remove(socio)
                print(f"Socio con email '{email}' dado de baja correctamente.")
                return
        print(f"No se encontró ningún socio con email '{email}'.")

    def darAltaPelicula(self, pelicula):
        if pelicula in self.peliculas:
            print(f"La película '{pelicula.titulo}' ya está en el catálogo.")
        else:
            self.peliculas.append(pelicula)
            print(f"Película '{pelicula.titulo}' añadida al catálogo correctamente.")

    def darBajaPelicula(self, titulo):
        for pelicula in self.peliculas:
            if pelicula.titulo.lower() == titulo.lower():
                self.peliculas.remove(pelicula)
                print(f"Película '{titulo}' eliminada del catálogo correctamente.")
                return
        print(f"No se encontró ninguna película con título '{titulo}'.")

    def listar_socios(self):
        if not self.socios:
            print("No hay socios registrados.")
        else:
            print("=== Lista de Socios ===")
            for i, socio in enumerate(self.socios, 1):
                print(f"  {i}. {socio}")

    def listar_peliculas(self):
        if not self.peliculas:
            print("No hay películas en el catálogo.")
        else:
            print("=== Catálogo de Películas ===")
            for i, pelicula in enumerate(self.peliculas, 1):
                print(f"  {i}. {pelicula}")

    def buscar_pelicula(self, titulo):
        for pelicula in self.peliculas:
            if pelicula.titulo.lower() == titulo.lower():
                print(f"Película encontrada: {pelicula}")
                return pelicula
        print(f"No se encontró ninguna película con título '{titulo}'.")
        return None

    def mostrar(self):
        return (f"Netflix | Socios: {len(self.socios)} | "
                f"Películas en catálogo: {len(self.peliculas)}")
```

```python
# --- Instancia de Netflix ---
netflix = Netflix()

# --- Crear y añadir películas ---
p1 = Pelicula("El Padrino", "Drama/Crimen", 1972, 175, "Francis Ford Coppola")
p2 = Pelicula("Inception", "Ciencia Ficción", 2010, 148, "Christopher Nolan")
p3 = Pelicula("El Rey León", "Animación", 1994, 88, "Roger Allers")

netflix.alta_pelicula(p1)
netflix.alta_pelicula(p2)
netflix.alta_pelicula(p3)
netflix.alta_pelicula(p1)  # Intento de duplicado

print()

# --- Crear y añadir socios ---
s1 = Socio("Ana", "García López", "ana.garcia@email.com")
s2 = Socio("Carlos", "Martínez Ruiz", "carlos.martinez@email.com")
s3 = Socio("Laura", "Sánchez Pérez", "laura.sanchez@email.com")

netflix.alta_socio(s1)
netflix.alta_socio(s2)
netflix.alta_socio(s3)
netflix.alta_socio(s2)  # Intento de duplicado

print()

# --- Listar socios y películas ---
netflix.listar_socios()
print()
netflix.listar_peliculas()
print()

# --- Buscar una película ---
netflix.buscar_pelicula("Inception")
netflix.buscar_pelicula("Titanic")  # No existe
print()

# --- Dar de baja ---
netflix.baja_pelicula("El Rey León")
netflix.baja_socio("carlos.martinez@email.com")
netflix.baja_socio("noexiste@email.com")  # No existe
print()

# --- Listar de nuevo tras las bajas ---
netflix.listar_socios()
print()
netflix.listar_peliculas()
print()

# --- Resumen del estado de Netflix ---
print(netflix)
```

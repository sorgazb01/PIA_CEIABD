# **Tienda Online**

En esta práctica se deberá crear un programa para **modelar una tienda online que venda diferentes tipos de productos.**

Para ello, se utilizaremos los conceptos vistos de **Programación Orientada a Objetos.**

### **Clase Producto**

Para empezar, se deberá crear una clase **abstracta llamada Producto,** que definirá los atributos y métodos que tendrán en común todos los productos que se vendan en la tienda. 

Esta clase abstracta deberá tener al menos los siguientes atributos:

* **nombre:** el nombre del producto.
* **precio:** el precio del producto.
* **descripcion:** una breve descripción del producto.
* **esta_oferta:** booleano para indicar si el producto está de oferta o no.

**Además, la clase Producto deberá tener al menos los siguientes métodos abstractos:**

* **calcular_precio:** un método que calculará el precio final del producto. Este método deberá ser implementado en las clases hijas de Producto.

* **mostrar_informacion:** un método que mostrará por pantalla la información básica del producto.



### Clases Hijas - Libro, Disco, Electronico

A continuación, se deberán crear **varias clases hijas de Producto para representar diferentes tipos de productos que se vendan en la tienda.**


* **Libro:** representa un libro que se venda en la tienda. 
    * Esta clase deberá tener el atributo adicional **autor**
    * Método para calcular el precio final que tendrá en cuenta si el libro está en oferta o no.

* **Disco:** representa un disco que se venda en la tienda. 
    * Esta clase deberá tener el atributo adicional **artista** 
    * Método para calcular el precio final que tendrá en cuenta si el disco está en oferta o no.

* **Electronico:** representa un producto electrónico que se venda en la tienda. 
    * Esta clase deberá tener un atributo adicional llamado **marca**
    * Un método para calcular el precio final que tendrá en cuenta si el producto está en oferta o no.

**Si lo consideras oportuno, cualquier otra clase hija que quieras implementar**


### Clase Tienda

Finalmente, se deberá crear una clase llamada **Tienda que represente la tienda online en sí misma.** 

Esta clase deberá tener al menos los siguientes métodos:

* **agregar_producto:** un método para agregar un producto a la tienda.

* **mostrar_productos:** un método para mostrar por pantalla todos los productos que se venden en la tienda.

* **buscar_productos:** un método para buscar productos por nombre o descripción.

* **calcular_precio_total:** un método que calculará el precio total de todos los productos en la tienda y lo mostrará por pantalla.

<font color='green'>**La clase Tienda deberá utilizar la herencia y el polimorfismo para manejar todos los productos que se agreguen a la tienda.**</font>


```python
from abc import ABC, abstractmethod

# Clase Abstracta Producto
class Producto(ABC):
    DESCUENTO_OFERTA = 0.20

    def __init__(self, nombre, precio, descripcion, esta_oferta=False):
        self.nombre = nombre
        self.precio = precio
        self.descripcion = descripcion
        self.esta_oferta = esta_oferta

    @abstractmethod
    def calcular_precio(self):
        pass

    @abstractmethod
    def mostrar_informacion(self):
        pass

# Clase Hija Libro
class Libro(Producto):
    def __init__(self, nombre, precio, descripcion, autor, esta_oferta=False):
        super().__init__(nombre, precio, descripcion, esta_oferta)
        self.autor = autor

    def calcular_precio(self):
        if self.esta_oferta:
            return self.precio * (1 - self.DESCUENTO_OFERTA)
        return self.precio

    def mostrar_informacion(self):
        return (f'Libro: {self.nombre}, Autor: {self.autor}, Precio: {self.calcular_precio():.2f}€, Descripción: {self.descripcion}')

# Clase Hija Disco
class Disco(Producto):
    def __init__(self, nombre, precio, descripcion, artista, esta_oferta=False):
        super().__init__(nombre, precio, descripcion, esta_oferta)
        self.artista = artista

    def calcular_precio(self):
        if self.esta_oferta:
            return self.precio * (1 - self.DESCUENTO_OFERTA)
        return self.precio

    def mostrar_informacion(self):
        return (f'Disco: {self.nombre}, Artista: {self.artista}, Precio: {self.calcular_precio():.2f}€, Descripción: {self.descripcion}')

# Clase Hija Electrónico
class Electronico(Producto):
    def __init__(self, nombre, precio, descripcion, marca, esta_oferta=False):
        super().__init__(nombre, precio, descripcion, esta_oferta)
        self.marca = marca

    def calcular_precio(self):
        if self.esta_oferta:
            return self.precio * (1 - self.DESCUENTO_OFERTA)
        return self.precio
    
    def mostrar_informacion(self):
        return (f'Electrónico: {self.nombre}, Marca: {self.marca}, Precio: {self.calcular_precio():.2f}€, Descripción: {self.descripcion}')

# Clase Tienda
class Tienda:
    def __init__(self, nombre):
        self.nombre = nombre
        self.productos = []

    def agregar_producto(self, producto):
        if not isinstance(producto, Producto):
            print('Error: solo se pueden agregar productos válidos.')
            return
        self.productos.append(producto)
        print(f'Producto {producto.nombre} añadido a la tienda.')

    def mostrar_productos(self):
        if not self.productos:
            print('La tienda no tiene productos.')
            return
        print(f'Catálogo de {self.nombre}')
        for i, producto in enumerate(self.productos, 1):
            print(f'{i}. {producto.mostrar_informacion()}')

    def buscar_productos(self, producto):
        producto = producto.lower()
        resultados = [p for p in self.productos
                      if producto in p.nombre.lower() or producto in p.descripcion.lower()]
        if not resultados:
            print(f'No se encontraron productos para "{producto}".')
        else:
            print(f'Resultados para {producto}:')
            for producto in resultados:
                print(f'{producto.mostrar_informacion()}')
        return resultados

    def calcular_precio_total(self):
        total = sum(p.calcular_precio() for p in self.productos)
        print(f'Precio total de todos los productos: {total:.2f}€')
        return total
```

```python
# Demostracion
tienda = Tienda('Tienda Online')

# Aniadir productos
tienda.agregar_producto(Libro('El Quijote', 12.99, 'Novela del caballero manchego Don Quijote y su fiel escudero Sancho Panza', 'Cervantes'))
tienda.agregar_producto(Libro('Clean Code', 35.00, 'Guia de buenas practicas para escribir codigo limpio y mantenible', 'Robert C. Martin', esta_oferta=True))
tienda.agregar_producto(Disco('Thriller', 15.50, 'Sexto album de estudio de Michael Jackson, publicado en 1982', 'Michael Jackson'))
tienda.agregar_producto(Disco('Back in Black', 13.00, 'Album de hard rock de AC/DC lanzado en 1980', 'AC/DC', esta_oferta=True))
tienda.agregar_producto(Electronico('iPhone 15', 999.00, 'Smartphone con chip A16, pantalla Super Retina XDR y camara de 48 MP', 'Apple'))
tienda.agregar_producto(Electronico('Galaxy S24', 849.00, 'Smartphone con pantalla Dynamic AMOLED y procesador Exynos 2400', 'Samsung', esta_oferta=True))

# Mostrar todos los productos
tienda.mostrar_productos()

# Buscar productos
tienda.buscar_productos('album')
tienda.buscar_productos('smartphone')

# Precio total
tienda.calcular_precio_total()
```

### Creando métodos que devuelvan un valor ordenable.

**Ahora imagina que tenemos una lista de personas y cada persona tiene como atributo una lista de objetos, cada objeto tiene un peso y queremos quedarnos con la persona con el objeto más pesado, como ordenamos con sorted?**

Piensa una solución que nos permita ordenar en base a ese criterio.

Comencemos creando una clase objeto básica:

- Nombre_objeto

- Peso

Una vez has creado la clase objeto:

- Modifica la clase persona para agregarle una lista de objetos

- Para cada persona crea un número aleatorio de objetos entre 1-5 con un peso aleatorio entre 1-10.

- Muestra la persona con el objeto más pesado

- Muestra la persona con más objetos.

```python
import random

class Objeto:
    def __init__(self, nombre_objeto, peso):
        self.nombre_objeto = nombre_objeto
        self.peso = peso

    def __str__(self):
        return f'{self.nombre_objeto} ({self.peso} kg)'


class Persona:
    def __init__(self, nombre, edad, altura):
        self.nombre = nombre
        self.edad = edad
        self.altura = altura
        self.objetos = [
            Objeto(f'Objeto {i+1}', random.randint(1, 10))
            for i in range(random.randint(1, 5))
        ]

    def objetoPesoMaximo(self):
        return max(self.objetos, key=lambda o: o.peso).peso

    def cantidadObjetos(self):
        return len(self.objetos)

    def __str__(self):
        objetos_str = ', '.join(str(o) for o in self.objetos)
        return f'{self.nombre} ({self.edad} años, {self.altura} cm) -> Objetos: [{objetos_str}]'

# Lista de Personas
personas = [
    Persona('Juan', 70, 180),
    Persona('María', 25, 170),
    Persona('Pedro', 30, 175),
    Persona('Antonio', 30, 200),
    Persona('Lucía', 22, 165)
]

print('Lista de personas y objetos')
for p in personas:
    print(p)

# Persona con el objeto más pesado
personaObjetoMasPesado = max(personas, key=lambda p: p.objetoPesoMaximo())
print(f'Persona con el objeto más pesado: {personaObjetoMasPesado.nombre} '
      f'(objeto más pesado: {personaObjetoMasPesado.objetoPesoMaximo()} kg)')

# Persona con más objetos
personaMasObjetos = max(personas, key=lambda p: p.cantidadObjetos())
print(f'Persona con más objetos: {personaMasObjetos.nombre} '
      f'({personaMasObjetos.cantidadObjetos()} objetos)')
```

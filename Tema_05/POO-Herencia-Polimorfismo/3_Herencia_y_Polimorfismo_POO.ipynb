## Herencia



* La herencia es un mecanismo de la programación orientada a objetos que **permite que una clase herede atributos y métodos de otra clase.** 


* De esta manera, **se pueden crear clases "hijas" que comparten ciertas características con una clase "padre"** y que pueden sobrescribir o añadir nuevas funcionalidades. 



* La sintaxis para definir una clase hija que hereda de otra clase es la siguiente:

```python
class Animal:
    
    def __init__(self, num_patas:int, color:str):
        self.patas = num_patas
        self.color = color
        
    
class Perro(Animal):# Ponemos entre parentesis las clases de la que hereda.
    
    def __init__(self, nombre:str, color:str, raza:str):
        Animal.__init__(self, 4, color)        
        self.nombre = nombre
        self.raza = raza
        
    
class Gato(Animal):# Ponemos entre parentesis las clases de la que hereda.
    
    def __init__(self, nombre:str, color:str, vidas:int, raza:str):
        #podemos llamar al constructor padre así tmb, pero cuando hay herencia multiple lo mejor es usar la opcion 
        #definida en Perro.
        super().__init__(num_patas=4, color=color)
        self.nombre = nombre
        self.vidas  = vidas
        self.raza   = raza
        
    
```

```python
animal1 = Animal(2,'rojo')
print(animal1.patas)
```

```python
bobi = Perro("Bobi", 'Negro', 'Pastor Aleman')
print(bobi.nombre)

manolin = Gato(nombre="Manolin", color='Pardo', vidas=4, raza='callejero')
print(manolin.nombre) 
```

```python
print(manolin)
```

**No se muestran los datos de los objetos porque necesitamos definir métodos para ello.**

### Sobrecarga de Métodos

* Los métodos sobrecargados **son métodos con el mismo nombre que se implementan en diferentes clases que heredan de una clase común.** 


* **Esto permite que cada clase tenga su propia implementación de ese método,** pero que se pueda acceder a ellos de manera uniforme. Un ejemplo común de un método sobrecargado es el **método "str", que se utiliza para convertir un objeto en una cadena de texto.**


* Por ejemplo, si tenemos una clase "Animal" que tiene un método "comer", podemos crear una clase "Perro" que herede de "Animal" y que tenga su propia implementación del método "comer". 


* De esta manera, un objeto "perro" tendrá acceso tanto al método "comer" de la clase "Animal" como al de la clase "Perro".

```python
class Animal:
    
    def __init__(self, num_patas:int, color:str):
        self.patas = num_patas
        self.color = color
    
    def comer(self):
        print("*** Método comer en Animal ***")
        
    def __str__(self):
        cadena = "Datos Comunes Animal:\n"
        cadena += f"Patas: {self.patas}\n"
        cadena += f"Color: {self.color}\n"        
        return cadena        
    
    
class Perro(Animal):
    
    def __init__(self, nombre:str, color:str, raza: str):
        Animal.__init__(self,4, color)
        self.nombre = nombre
        self.raza   = raza
        
    def comer(self):    
        print("*** Método comer en Perro ***")
        print(f"El perro {self.nombre} come pienso")
        
        
    def __str__(self):
        cadena = Animal.__str__(self)
        cadena += "Datos de Perro:\n"
        cadena += f"Nombre: {self.nombre}\n"
        cadena += f"Raza: {self.raza}"
        return cadena                   
    
class Gato(Animal):
    
    def __init__(self, nombre:str, color:str, vidas:int, raza:str):
        #podemos llamar al constructor padre así tmb, pero cuando hay herencia multiple lo mejor es usar la opcion 
        #definida en Perro.
        super().__init__(num_patas=4, color=color)
        self.nombre = nombre
        self.vidas  = vidas
        self.raza   = raza
        
    def comer(self):
        print("*** Método comer en Gato ***")
        print(f"El gato {self.nombre} come palomas")
        
        
    def __str__(self):
        cadena = Animal.__str__(self)
        cadena += "Datos de Gato:\n"
        cadena += f"Nombre: {self.nombre}\n"
        cadena += f"Raza: {self.raza}\n"
        cadena += f"Vidas: {self.vidas}"
        return cadena            
        
```

```python
bobi = Perro(nombre="Bobi", color='Negro', raza='Pastor Aleman')
print(bobi)
```

```python
manolin = Gato(nombre="Manolin",color='Pardo', vidas=4, raza='callejero')
print(manolin)
```

Como vemos, al imprimir los datos del perro y del gato, ahora se imprimen de forma automática los datos de cada una de las subclases. **Esto es posible a la sobrecarga del método str**

## Polimorfismo en Python

* El polimorfismo en Python se refiere a la **capacidad de un objeto de una clase derivada para comportarse de maneras diferentes en función de la situación.** 


* Esto se logra mediante la **sobrecarga de métodos y la herencia,** de manera que un mismo método puede tener diferentes implementaciones en diferentes clases.

```python
bobi.comer()
manolin.comer()
```

**<div class="alert alert-info"> El mismo método hace que los animales se comporten de una forma u otra dependiendo del subtipo de animal que sean</div>**

### Duck Typing


Duck typing es una **filosofía de programación** que se basa en el principio:
>**"Si camina como un pato y grazna como un pato, entonces es un pato".**

En otras palabras, **en lugar de preocuparse por el tipo de un objeto en tiempo de compilación, el código espera que el objeto tenga ciertos métodos y propiedades** y confía en que el objeto sabrá cómo manejarlos.

En pocas palabras, **a Python le dan igual los tipos de los objetos, lo único que le importan son los métodos.**




### Utilidades de este enfoque

* En Python, esto se traduce en un **enfoque más flexible para la definición de clases y la interacción entre ellas.** 
    
    * Por ejemplo, en lugar de tener que definir una jerarquía de clases y hacer que una clase herede de otra para poder interactuar con ella, en Python puedes simplemente **definir una clase que implemente los métodos y propiedades que necesitas,** y luego usarla como si fuera una instancia de cualquier otra clase que implemente esos mismos métodos y propiedades.
    
    

* Esto **puede ser útil** en situaciones en las que **no sabes de antemano qué tipo de objetos tendrás que tratar** en tu código, o cuando quieres que tu código sea lo más **genérico posible y pueda manejar diferentes tipos de objetos de manera transparente.** De esta manera, puedes escribir código que sea más **fácil de mantener y más fácil de adaptar** a diferentes situaciones.





**Veamos un ejemplo**

Definamos tres clases de animales distintas que implementan el método hablar(). Nótese que no existe herencia entre ellas, son clases totalmente independientes. De haberla estaríamos hablando de polimorfismo.

```python
class Perro:
    def hablar(self):
        print("¡Guau, Guau!")

class Gato:
    def hablar(self):
        print("¡Miau, Miau!")

class Vaca:
    def hablar(self):
        print("¡Muuu, Muuu!")
        
def llama_hablar(x):
    x.hablar()        
```

Como puedes observar, en Python **no es necesario especificar los tipos,** simplemente decimos que el parámetro de entrada tiene el nombre x, pero no especificamos su tipo.

Cuando Python entra en la función y evalúa `x.hablar()`, le da igual el tipo al que pertenezca x siempre y cuando tenga el método `hablar()`. Esto es el **duck typing en todo su esplendor.**

```python
lista = [Perro(), Gato(), Vaca()]
for animal in lista:
    animal.hablar()
```

## Ejemplo Herencia y Polimorfismo

Un ejemplo avanzado de programa con varias clases que implementen herencia y polimorfismo podría ser un sistema de gestión de empleados de una empresa. Podríamos tener una clase "Empleado" que define atributos comunes a todos los empleados, como el nombre y la edad. Luego, podríamos tener clases "Gerente" y "Operario" que hereden de "Empleado" y que tengan atributos y métodos específicos de cada tipo de empleado. Además, podríamos tener una clase "Empresa" que tenga una lista de empleados y que tenga un método "contratar" que pueda agregar empleados de cualquier tipo a esa lista, gracias al polimorfismo. La estructura del código podría ser la siguiente:

```python
class Empleado:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
        
    def trabajar(self):
        print(f"El empleado {self.nombre} trabaja")

class Gerente(Empleado):
    def __init__(self, nombre, edad, departamento):
        super().__init__(nombre, edad)
        self.departamento = departamento
    def trabajar(self):
        print(f"El empleado {self.nombre} Dirige a sus gerenciados")

class Operario(Empleado):
    def __init__(self, nombre, edad, herramienta):
        super().__init__(nombre, edad)
        self.herramienta = herramienta
        
    def trabajar(self):
        print(f"El empleado {self.nombre} utiliza la herramienta {self.herramienta}")        

class Empresa:
    def __init__(self):
        self.empleados = []#lista de empleados

    def contratar(self, empleado):
        #agregar empleados de cualquier tipo
        self.empleados.append(empleado)
        
    def jornada_trabajo(self):
        for trabajador in self.empleados:
            trabajador.trabajar()

```

```python
empresa = Empresa()

gerente = Gerente("Juan", 30, "Ventas")
operario = Operario("Ana", 25, "Martillo")
empleado = Empleado("Alexa", 45)

empresa.contratar(gerente)
empresa.contratar(operario)
empresa.contratar(empleado)

```

```python
empresa.jornada_trabajo()
```

En este ejemplo, la clase "Empresa" no tiene conocimiento de la clase de los empleados que se están contratando, pero puede agregarlos a su lista de empleados gracias al polimorfismo.

## Herencia Multiple

La herencia múltiple en Python se refiere a la capacidad de una clase de heredar atributos y métodos de más de una clase padre. Esto puede ser útil si una clase necesita heredar características de varias clases diferentes, en lugar de solo de una clase padre.

Para implementar la herencia múltiple en Python, se utiliza la palabra clave class seguida del nombre de la clase hija, seguida de una tupla que contiene los nombres de las clases padres entre paréntesis y separados por comas. Por ejemplo:

```python
class ClaseHija(ClasePadre1, ClasePadre2):
    # código de la clase hija aquí
```

La clase hija heredará los atributos y métodos de ambas clases padre, lo que le permite utilizarlos como si fueran suyos propios. Por ejemplo, si tenemos dos clases padre llamadas Persona y Empleado, y queremos crear una clase hija llamada Programador que herede de ambas clases, podríamos hacerlo de la siguiente manera:

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad
    
    def saludar(self):
        print(f"Hola, mi nombre es {self.nombre} y tengo {self.edad} años.")

class Empleado:
    def __init__(self, salario):
        self.salario = salario
    
    def mostrar_salario(self):
        print(f"Mi salario es de {self.salario}.")

class Programador(Persona, Empleado):
    def __init__(self, nombre, edad, salario):
        # Llamamos al método __init__ de la clase padre Persona
        Persona.__init__(self, nombre, edad)
        
        # Llamamos al método __init__ de la clase padre Empleado
        Empleado.__init__(self, salario)
        #podemos utilizar tambien el método super(). pero con herencia múltiple no es recomendable

        



```

```python
# Creamos una instancia de la clase Programador
p = Programador("Juan", 28, 30000)
```

```python
# Podemos usar los métodos de la clase Persona
p.saludar()

# También podemos usar los métodos de la clase Empleado
p.mostrar_salario()
```

## Utilizando Métodos Mágicos.

### Ordenando Objetos por un criterio

Puedes usar la función **max o sorted** en Python para encontrar el máximo u ordenar una lista de objetos. **Sin embargo, la función max requiere que los objetos en la lista sean comparables entre sí.** Esto significa que **deben tener una forma de comparar dos objetos de ese tipo para determinar cuál es "mayor" que el otro.**

Para hacer esto con una clase Persona, necesitarás definir un método \__lt__ en tu clase Persona que **permita a Python comparar dos objetos de tipo Persona y determinar cuál es "menor" entre los dos.** 

Una vez que hayas definido este método, puedes usar la función max para encontrar la persona más "grande" en tu lista.

Por ejemplo, si tienes una clase Persona que tiene un atributo edad:

```python
class Persona:
    def __init__(self, nombre, edad):
        self.nombre = nombre
        self.edad = edad

    def __lt__(self, otro):
        return self.edad < otro.edad
    
    def __str__(self):
        return f"Nombre: {self.nombre} Edad: ({self.edad} años)"    
```

Puedes usar la función max para encontrar la persona más grande en una lista de objetos Persona de la siguiente manera:

```python
personas = [
  Persona("Juan", 70),
  Persona("María", 25),
  Persona("Pedro", 30)
]

persona_mas_grande = max(personas)

print(f"La persona más grande es {persona_mas_grande.nombre} con {persona_mas_grande.edad} años")

```

```python
print(f"La persona más grande es: {persona_mas_grande}")
```

```python
personas_ordenadas = sorted(personas)
for persona in personas:
    print(persona)
```

Ahora puedo usar sorted para ordenar la lista de personas, de forma automática está cogiendo el método \__lt__ para ordenar las personas

### Varios Criterios de Ordenación

¿ Y podria tener varios criterios para ordenar?

Por ejemplo poder comparar por edad o poder comparar por la altura de una persona

Sí, es posible tener varios métodos de comparación en una clase en Python. Para hacer esto, puedes definir varios métodos en la clase que se encarguen de comparar los objetos de ese tipo de diferentes maneras.

Por ejemplo, si quieres que tu clase Persona se pueda comparar por edad y por altura, puedes definir los métodos \__lt__ y \__gt__ de la siguiente manera:

```python
class Persona:
    def __init__(self, nombre, edad, altura):
        self.nombre = nombre
        self.edad = edad
        self.altura = altura

    def __lt__(self, otro):
        # Compara por edad si ambos tienen la misma altura
        if self.altura == otro.altura:
            return self.edad < otro.edad
        # Compara por altura si tienen diferente altura
        else:
            return self.altura < otro.altura

    def __gt__(self, otro):
        # Compara por edad si ambos tienen la misma altura
        if self.altura == otro.altura:
            return self.edad > otro.edad
        # Compara por altura si tienen diferente altura
        else:
            return self.altura > otro.altura
```

Con esta implementación, cuando llames a la función max con una lista de Persona, se devolverá la persona con la edad más grande si tienen la misma altura, o la persona con la altura más grande si tienen diferente altura.

## Sorted + Lambdas

Vamos a estudiar ahora otro método de comparación de objetos, combinando el método sorted con funciones lambdas.

La función sorted te **permite especificar una función de comparación personalizada que se usará para ordenar la lista.**

Comencemos definiendo la clase persona.

### Utilizando atributos que pueden ordenarse

```python
class Persona:
    def __init__(self, nombre, edad, altura):
        self.nombre = nombre
        self.edad = edad
        self.altura = altura

    def __str__(self):
        return f"{self.nombre} ({self.edad} años, {self.altura} cm)"

```

Por ejemplo, si quieres ordenar una lista de Persona por edad y luego por altura, puedes hacerlo de la siguiente manera:

```python
personas = [
  Persona("Juan", 70, 180),
  Persona("María", 25, 170),
  Persona("Pedro", 30, 170),
  Persona("Antonio", 30, 200)
]

# Ordena la lista por edad y luego por altura
ordenada_por_edad_altura = sorted(personas, key=lambda p: (p.edad, p.altura))

# Imprime la lista ordenada ascendentemente
print("Lista ordenada por edad y luego por altura:")
for persona in ordenada_por_edad_altura:
    print(persona)

# Ordena la lista por altura y luego por edad ascendentemente
ordenada_por_altura_edad = sorted(personas, key=lambda p: (p.altura, p.edad))

# Imprime la lista ordenada
print("Lista ordenada por altura y luego por edad:")
for persona in ordenada_por_altura_edad:
    print(persona)

```

### Creando métodos que devuelvan un valor ordenable.

**Ahora imagina que tenemos una lista de personas y cada persona tiene como atributo una lista de objetos, cada objeto tiene un peso y queremos quedarnos con la persona con el objeto más pesado, como ordenamos con sorted?**

Piensa una solución que nos permita ordenar en base a ese criterio.

Comencemos creando una clase objeto básica:

- Nombre_objeto

- Peso

Una vez has creado la clase objeto:

- Modifica la clase persona para agregarle una lista de objetos

- Para cada persona crea un número aleatorio de objetos entre 1-5 con un peso aleatorio entre 1-10.

- Muestra la persona con el objeto más pesado

- Muestra la persona con más objetos.


```python
import random

class Objeto:
    def __init__(self, nombre_objeto, peso):
        self.nombre_objeto = nombre_objeto
        self.peso = peso

    def __str__(self):
        return f"{self.nombre_objeto} ({self.peso} kg)"


class Persona:
    def __init__(self, nombre, edad, altura):
        self.nombre = nombre
        self.edad = edad
        self.altura = altura
        self.objetos = [
            Objeto(f"Objeto_{i+1}", random.randint(1, 10))
            for i in range(random.randint(1, 5))
        ]

    def peso_maximo_objeto(self):
        return max(self.objetos, key=lambda o: o.peso).peso

    def cantidad_objetos(self):
        return len(self.objetos)

    def __str__(self):
        objetos_str = ", ".join(str(o) for o in self.objetos)
        return f"{self.nombre} ({self.edad} años, {self.altura} cm) -> Objetos: [{objetos_str}]"


# Crear lista de personas
personas = [
    Persona("Juan", 70, 180),
    Persona("María", 25, 170),
    Persona("Pedro", 30, 175),
    Persona("Antonio", 30, 200),
    Persona("Lucía", 22, 165),
]

print("=== Lista de personas con sus objetos ===")
for p in personas:
    print(p)

# Persona con el objeto más pesado
persona_objeto_pesado = max(personas, key=lambda p: p.peso_maximo_objeto())
print(f"\nPersona con el objeto más pesado: {persona_objeto_pesado.nombre} "
      f"(objeto más pesado: {persona_objeto_pesado.peso_maximo_objeto()} kg)")

# Persona con más objetos
persona_mas_objetos = max(personas, key=lambda p: p.cantidad_objetos())
print(f"Persona con más objetos: {persona_mas_objetos.nombre} "
      f"({persona_mas_objetos.cantidad_objetos()} objetos)")
```
